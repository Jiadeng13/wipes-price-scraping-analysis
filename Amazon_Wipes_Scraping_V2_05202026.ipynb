````python
import pandas as pd
import asyncio
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError
from bs4 import BeautifulSoup
import re
import os

# --- Helper functions (adapted from previous code for consistency) ---
def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()

def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except ValueError:
        return None

def format_price_per_count(value: float) -> str:
    return f"${value:.4f}/count"

# --- New Scraping Function: Scrape prices directly from Amazon product page via Google ---
async def scrape_amazon_prices_from_google_redirect(asin: str, product_title: str) -> tuple[str, str]:
    """
    Searches Google for an ASIN, navigates to the Amazon product page, and extracts CurrentPrice and PricePerCount.
    """
    search_query = f"amazon.com {asin}" # More direct search for Amazon product
    print(f"Searching Google for ASIN: '{asin}' (Product: {product_title[:50]}...)")

    current_price = "N/A"
    price_per_count = "N/A"

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox"])
        context = await browser.new_context(
            user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/123.0.0.0 Safari/537.36"
        )
        page = await context.new_page()
        page.set_default_timeout(90000) # Increased general timeout to 90 seconds

        try:
            await page.goto(f"https://www.google.com/search?q={search_query}", wait_until="domcontentloaded")
            await page.wait_for_selector("#search", timeout=60000) # Wait for Google search results

            # Find the Amazon product link for the specific ASIN
            # Look for links containing amazon.com/dp/ASIN or amazon.com/ASIN/dp
            amazon_link_selector = f"a[href*='amazon.com/dp/{asin}'], a[href*='amazon.com/{asin}/dp']"
            amazon_product_link = await page.query_selector(amazon_link_selector)

            if amazon_product_link:
                href = await amazon_product_link.get_attribute('href')
                if href:
                    print(f"Found Amazon link for ASIN {asin}: {href}")
                    await page.goto(href, wait_until="domcontentloaded")
                    await page.wait_for_selector("#corePriceDisplay_desktop_feature_div, #priceblock_ourprice, .a-price", timeout=60000) # Wait for price element on Amazon page

                    # Extract CurrentPrice
                    current_price_element = await page.query_selector("#corePriceDisplay_desktop_feature_div .a-price-whole, #priceblock_ourprice, .a-price-whole, .a-offscreen")
                    if current_price_element:
                        text_content = await current_price_element.inner_text()
                        price_match = re.search(r'\$\d+\.\d{2}', text_content)
                        if price_match:
                            current_price = price_match.group(0).strip()
                        else:
                            # Fallback for hidden offscreen spans that contain the full price string
                            offscreen_price_element = await page.query_selector("span.a-price.a-offscreen, span#priceblock_ourprice.a-size-medium.a-color-price")
                            if offscreen_price_element:
                                current_price = (await offscreen_price_element.inner_text()).strip()
                            elif "a-price-whole" in await current_price_element.evaluate('el => el.className'): # Check if it's the whole part of a price
                                fraction_element = await page.query_selector(".a-price-fraction")
                                if fraction_element:
                                    current_price = (await current_price_element.inner_text()).strip() + (await fraction_element.inner_text()).strip()
                                    if not current_price.startswith('$'):
                                        current_price = '$' + current_price

                    # Extract PricePerCount
                    price_per_count_element = await page.query_selector("#corePriceDisplay_desktop_feature_div .a-size-base.a-color-secondary, span.a-size-base.a-color-secondary.a-text-normal")
                    if price_per_count_element:
                        ppc_text = (await price_per_count_element.inner_text()).strip()
                        if '/count' in ppc_text or '/unit' in ppc_text or 'per' in ppc_text: # Refine extraction to ensure it's a unit price
                             # Try to capture something like '$X.XX / Y' or '($X.XX/count)'
                            ppc_match = re.search(r'\(\$.*?(?:/count|per.*)\)', ppc_text, re.IGNORECASE)
                            if ppc_match:
                                price_per_count = ppc_match.group(0).strip('() ')
                            else:
                                price_per_count = ppc_text # Use raw text if complex regex fails

            else:
                print(f"Amazon product link for ASIN {asin} not found in Google results.")

        except PlaywrightTimeoutError:
            print(f"Timeout during scraping for ASIN {asin}.")
        except Exception as e:
            print(f"Error during scraping for ASIN {asin}: {e}")
        finally:
            await browser.close()

    return current_price, price_per_count

# --- Main function to update missing prices ---
async def update_missing_amazon_prices():
    input_csv_path = "/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/amazon_adult_wipes_detailed.csv" # Use the local output path from previous runs
    output_csv_path = "/content/amazon_adult_wipes_detailed_updated_prices.csv" # Save to a new file to preserve original

    if not os.path.exists(input_csv_path):
        print(f"Error: Input file not found at '{input_csv_path}'. Please ensure the Amazon scraping ran correctly and outputted to /content/.")
        return

    df = pd.read_csv(input_csv_path)

    # Ensure required columns exist
    required_columns = ['ProductTitle', 'ASIN', 'CurrentPrice', 'PricePerCount']
    if not all(col in df.columns for col in required_columns):
        print(f"Error: Missing one or more required columns ({required_columns}) in the input CSV. Found: {df.columns.tolist()}")
        return

    # Filter for rows where CurrentPrice is 'N/A' or empty, or NaN
    # Convert 'CurrentPrice' to string to handle 'N/A' and empty strings, then check for NaN for actual missing values
    df['CurrentPrice'] = df['CurrentPrice'].astype(str).replace('', pd.NA).replace('N/A', pd.NA)
    missing_price_df = df[df['CurrentPrice'].isna()].copy()

    if missing_price_df.empty:
        print("No products with missing CurrentPrice found. No updates needed.")
        df.to_csv(output_csv_path, index=False) # Save original to new path
        print(f"Original DataFrame with no missing prices saved to: {output_csv_path}")
        return

    print(f"Found {len(missing_price_df)} products with missing CurrentPrice. Attempting to update...")

    updated_prices = []
    for index, row in missing_price_df.iterrows():
        product_title = row['ProductTitle']
        asin = row['ASIN']
        # Call the new scraping function
        current_price, price_per_count = await scrape_amazon_prices_from_google_redirect(asin, product_title)

        # Update the original DataFrame (df) directly
        df.loc[df['ASIN'] == asin, 'CurrentPrice'] = current_price
        df.loc[df['ASIN'] == asin, 'PricePerCount'] = price_per_count

        updated_prices.append({
            'ASIN': asin,
            'ProductTitle': product_title,
            'UpdatedCurrentPrice': current_price,
            'UpdatedPricePerCount': price_per_count
        })

    # Save the updated DataFrame
    df.to_csv(output_csv_path, index=False)
    print(f"\nUpdated DataFrame saved to: {output_csv_path}")

    # Display updated information for verification
    updated_df = pd.DataFrame(updated_prices)
    print("\nSummary of updated prices:")
    print(updated_df)

    # Optionally download the updated file
    from google.colab import files
    try:
        files.download(output_csv_path)
        print("Updated file downloaded successfully.")
    except Exception as e:
        print(f"Could not download the file: {e}")

# Run the main function
await update_missing_amazon_prices()
````

````python
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError


# ── HELPERS ──────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(r"[\d.]+", rating_span.get_text(strip=True)):
            rating = rating_span.get_text(strip=True)

    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


# ── SCRAPER ENGINE ────────────────

def build_search_urls(query: str) -> list[str]:
    """Builds the same price-sliced URL list for any query with $2 increments."""
    q = query.replace(" ", "+")
    urls = []

    # Price bracket: Less than $5.00
    urls.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A-500")

    # Price brackets with $2 increments: $5.00 - $7.00, $7.00 - $9.00, etc.
    # Start from $5 (500 cents) up to $101 (10100 cents), step is $2 (200 cents)
    for i in range(500, 10100, 200):
        urls.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A{i}-{(i+200)}")

    # Price bracket: $101.00 and above
    urls.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A10100-99999999") # Up to a very high number

    return urls

async def close_amazon_modals(page):
    """
    Tries to close Amazon redirect / sign-in / region modals if they appear.
    Safe to call even if no modal is present.
    """
    selectors = [
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            await page.wait_for_selector(sel, timeout=3000)
            await page.click(sel)
            # small pause after closing
            await page.wait_for_timeout(500)
            # once we close one, we can stop
            break
        except PlaywrightTimeoutError:
            # this particular selector didn't appear; try the next one
            continue
        except Exception:
            continue

async def run_category_scrape(query: str, output_path: str, max_pages_per_slice: int = 10):
    search_urls = build_search_urls(query)
    output_csv = Path(output_path)

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--no-sandbox"])
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in search_urls:
            print(f"Starting slice: {base_url}")
            await page.goto(base_url, wait_until="domcontentloaded", timeout=90000)

            for page_idx in range(1, max_pages_per_slice + 1):
                print(f"  Scraping page {page_idx}...")
                for _ in range(3):
                    await page.mouse.wheel(0, 1500)
                    await page.wait_for_timeout(1000)

                content = await page.content()
                soup = BeautifulSoup(content, "html.parser")
                items = soup.select("[data-component-type='s-search-result']")

                for item in items:
                    data = parse_item_detailed(item)
                    if data and data["ASIN"] not in seen_asins:
                        data["Page"] = page_idx
                        all_results.append(data)
                        seen_asins.add(data["ASIN"])

                next_btn = await page.query_selector("a.s-pagination-next")
                if next_btn:
                    await next_btn.click()
                    await page.wait_for_timeout(random.randint(3000, 5000))
                else:
                    break

        if all_results:
            keys = all_results[0].keys()
            with open(output_csv, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {output_csv}")

        await browser.close()
````

````python
# Amazon baby wipes https://www.amazon.com/s?k=baby+wipes
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

# SEARCH_URLS = (
#     [
#         "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A-500",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A500-600",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A600-700",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A700-800",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A800-900",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A900-1000",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1000-1100",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1100-1200",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1200-1300",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1300-1400",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1400-1500",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1500-1600",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1600-1700",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1700-1800",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1800-1900",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1900-2000",
# ]
#  + [
#     f"https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
#     for i in range(20, 105)
#  ]
#  + [
#      "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A10500-999999",  # $105 and above
#  ]
# )
SEARCH_URLS = (
  [
"https://www.amazon.com/s?k=%22B07YXNCYLW%22"
 ]
)

OUTPUT_CSV = Path("/content/amazon_honest_wipes_test.csv")
# OUTPUT_CSV = Path("/content/amazon_baby_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""

#72 Count (Pack of 8)
def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    print("extract_count_pack_from_text", text)
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                print("1", str(per), str(total // per if per else 1))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                print("2", str(total_val // case_val), str(case_val))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\\b", t, re.I)
    if simple_total:
        print("3", simple_total.group(1).replace(",", ""), "1")
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
# Amazon baby wipes https://www.amazon.com/s?k=baby+wipes
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

# SEARCH_URLS = (
#     [
#         "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A-500",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A500-600",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A600-700",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A700-800",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A800-900",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A900-1000",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1000-1100",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1100-1200",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1200-1300",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1300-1400",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1400-1500",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1500-1600",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1600-1700",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1700-1800",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1800-1900",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1900-2000",
# ]
#  + [
#     f"https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
#     for i in range(20, 105)
#  ]
#  + [
#      "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A10500-999999",  # $105 and above
#  ]
# )
SEARCH_URLS = (
  [
"https://www.amazon.com/s?k=%22B07YXNCYLW%22"
 ]
)

OUTPUT_CSV = Path("/content/amazon_honest_wipes_test.csv")
# OUTPUT_CSV = Path("/content/amazon_baby_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""

#
def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    print("extract_count_pack_from_text", text)
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                print("1", str(per), str(total // per if per else 1))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                print("2", str(total_val // case_val), str(case_val))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\\b", t, re.I)
    if simple_total:
        print("3", simple_total.group(1).replace(",", ""), "1")
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
# Amazon baby wipes https://www.amazon.com/s?k=baby+wipes
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

# SEARCH_URLS = (
#     [
#         "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A-500",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A500-600",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A600-700",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A700-800",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A800-900",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A900-1000",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1000-1100",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1100-1200",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1200-1300",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1300-1400",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1400-1500",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1500-1600",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1600-1700",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1700-1800",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1800-1900",
#     "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A1900-2000",
# ]
#  + [
#     f"https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
#     for i in range(20, 105)
#  ]
#  + [
#      "https://www.amazon.com/s?k=baby+wipes&rh=p_36%3A10500-999999",  # $105 and above
#  ]
# )
SEARCH_URLS = (
  [
"https://www.amazon.com/s?k=%22B07YXNCYLW%22"
 ]
)

OUTPUT_CSV = Path("/content/amazon_honest_wipes_test.csv")
# OUTPUT_CSV = Path("/content/amazon_baby_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""

#72 Count (Pack of 8)
def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    print("extract_count_pack_from_text", text)
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                print("1", str(per), str(total // per if per else 1))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                print("2", str(total_val // case_val), str(case_val))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\\b", t, re.I)
    if simple_total:
        print("3", simple_total.group(1).replace(",", ""), "1")
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
#Correct Count Pack Extract from Claude

import re

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()

def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        # (pattern, divide_flag)
        # divide=False: group(1)=count, group(2)=pack — return as-is
        # divide=True:  need arithmetic to derive one value from the other
        (r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?(?:Case|Pack)\s+of\s+(\d+)\s*\)?", False),  # Pattern 0
        (r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",           True),   # Pattern 1
        (r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",                         False),  # Pattern 2
        (r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",                                      False),  # Pattern 3
        (r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack", False),  # Pattern 4
        (r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",             True),   # Pattern 5
        (r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",                             False),  # Pattern 6
        (r"(\d+)\s+packs?\s+of\s+(\d+)",                                                 False),  # Pattern 7
    ]

    for pattern, divide in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if divide:
                if "Case" in pattern:  # Pattern 1: case_count × per_pack
                    case_val = int(m.group(1))
                    per_pack = int(m.group(2))
                    return str(per_pack), str(case_val)
                else:                  # Pattern 5: per_pack, total
                    per = int(m.group(1))
                    total = int(m.group(2))
                    return str(per), str(total // per if per else 1)
            else:
                return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""

# ── Test script ──────────────────────────────────────────────────────────────
input_string = "72 Count (Pack of 8)"
count_output, pack_output = extract_count_pack_from_text(input_string)

print(f"Input: '{input_string}'")
print(f"Expected Count = 72, Pack = 8")
print(f"Actual Output: Count = {count_output}, Pack = {pack_output}")
print(f"{'PASS' if count_output == '72' and pack_output == '8' else 'FAIL'}")
````

````python
# Amazon baby wipes https://www.amazon.com/s?k=flushable+wipes
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

SEARCH_URLS = (
    [
        "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A-500",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A500-600",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A600-700",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A700-800",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A800-900",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A900-1000",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1000-1100",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1100-1200",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1200-1300",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1300-1400",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1400-1500",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1500-1600",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1600-1700",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1700-1800",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1800-1900",
    "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A1900-2000",
]
#  + [
#     f"https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
#     for i in range(20, 105)
#  ]
 + [
     "https://www.amazon.com/s?k=flushable+wipes&rh=p_36%3A10500-999999",  # $105 and above
]
)

OUTPUT_CSV = Path("/content/amazon_flushable_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        # "576 Count (Case of 12)" or "Pack of 10"
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?\s*(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        # "Case of 12 Packs, 50 Count Per Pack"
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(\s*pack\s+of\s+(\d+)\s*\)",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
# Amazon baby wipes https://www.amazon.com/s?k=xl+wipes
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

SEARCH_URLS = (
    [
        "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A-500",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A500-600",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A600-700",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A700-800",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A800-900",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A900-1000",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1000-1100",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1100-1200",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1200-1300",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1300-1400",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1400-1500",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1500-1600",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1600-1700",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1700-1800",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1800-1900",
    "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A1900-2000",
]
 + [
    f"https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
    for i in range(20, 105)
 ]
 + [
     "https://www.amazon.com/s?k=xl+wipes&rh=p_36%3A10500-999999",  # $105 and above
 ]
)

OUTPUT_CSV = Path("/content/amazon_xl_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        # "576 Count (Case of 12)" or "Pack of 10"
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?\s*(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        # "Case of 12 Packs, 50 Count Per Pack"
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(\s*pack\s+of\s+(\d+)\s*\)",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
#Amazon baby wipes - Google Gemini supported (Doesn't scrape all the items)

import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError


# ── CONFIGURATION ───────────────────────

SEARCH_URLS = []
q = "baby+wipes"

# Price bracket: Less than $5.00 (500 cents)
SEARCH_URLS.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A-500")

# Price brackets: $5.00 - $7.00, $7.00 - $9.00, $9.00 - $11.00
SEARCH_URLS.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A500-700")
SEARCH_URLS.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A700-900")
SEARCH_URLS.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A900-1100")

# Price brackets with $2 increments: $11.00 - $13.00, ..., up to $59.00 - $61.00
# Start from $11 (1100 cents) up to $59 (5900 cents), step is $2 (200 cents)
for i in range(1100, 6000, 200):
    SEARCH_URLS.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A{i}-{(i+200)}")

# Price bracket: $60.00 and above
SEARCH_URLS.append(f"https://www.amazon.com/s?k={q}&rh=p_36%3A6000-99999999")


OUTPUT_CSV = Path("/content/amazon_baby_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        # "576 Count (Case of 12)" or "Pack of 10"
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?\s*(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        # "Case of 12 Packs, 50 Count Per Pack"
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ───────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page) # Call to dismiss modal after pagination
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=page.default_navigation_timeout)
                            await page.wait_for_load_state('networkidle', timeout=page.default_navigation_timeout) # Wait for the new page to load
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page) # Added to dismiss modal after next page loads
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
# Amazon adult wipes https://www.amazon.com/s?k=adult+wipes
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

SEARCH_URLS = (
    [
        "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A-500"
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A500-600",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A600-700",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A700-800",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A800-900",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A900-1000",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1000-1100",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1100-1200",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1200-1300",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1300-1400",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1400-1500",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1500-1600",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1600-1700",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1700-1800",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1800-1900",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1900-2000",
]
 + [
    f"https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
    for i in range(20, 105)
 ]
 + [
     "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A10500-999999",  # $105 and above
 ]
)

OUTPUT_CSV = Path("/content/amazon_adult_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        # "576 Count (Case of 12)" or "Pack of 10"
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?\s*(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        # "Case of 12 Packs, 50 Count Per Pack"
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(?(?:pack\s+of\s+(\d+))\s*\)?",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
import asyncio
import csv
import re
import random
from pathlib import Path
from bs4 import BeautifulSoup
from playwright.async_api import async_playwright, TimeoutError as PlaywrightTimeoutError

# ── CONFIGURATION ───────────────────────────────────────────────────────

SEARCH_URLS = (
    [
        "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A-500"
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A500-600",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A600-700",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A700-800",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A800-900",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A900-1000",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1000-1100",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1100-1200",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1200-1300",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1300-1400",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1400-1500",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1500-1600",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1600-1700",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1700-1800",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1800-1900",
    "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A1900-2000",
]
 + [
    f"https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A{i*100}-{(i+1)*100}"
    for i in range(20, 105)
 ]
 + [
     "https://www.amazon.com/s?k=adult+wipes&rh=p_36%3A10500-999999",  # $105 and above
 ]
)

OUTPUT_CSV = Path("/content/amazon_adult_wipes_detailed.csv")
MAX_PAGES_PER_SLICE = 10


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except Exception:
        return None


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        # "576 Count (Case of 12)" or "Pack of 10"
        r"(\d[\d,]*)\s+(?:Count|Wipes|Sheets)\s*\(?\s*(?:Case|Pack)\s+of\s+(\d+)\s*\)?",
        # "Case of 12 Packs, 50 Count Per Pack"
        r"Case\s+of\s+(\d+)\s+Packs?.*(\d+)\s+(?:Count|Wipes)\s+per\s+pack",
        r"(\d[\d,]*)\s+count\s*\(\s*pack\s+of\s+(\d+)\s*\)",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top|resealable|soft|travel)?\s*packs?\s*[,;]?\s*(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack",
        r"(\d+)\s+(?:wipes|count|sheets)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    for pattern in checks:
        m = re.search(pattern, t, re.I)
        if m:
            if "total" in pattern:
                per, total = int(m.group(1)), int(m.group(2))
                return str(per), str(total // per if per else 1)
            if "Case" in pattern and m.lastindex == 2:
                total_val = int(m.group(1).replace(",", ""))
                case_val = int(m.group(2).replace(",", ""))
                return str(total_val // case_val), str(case_val)
            return m.group(1).replace(",", ""), m.group(2).replace(",", "")

    simple_total = re.search(r"(\d[\d,]*)\s+(?:Count|Pieces|Wipes)\b", t, re.I)
    if simple_total:
        return simple_total.group(1).replace(",", ""), "1"

    return "", ""


def parse_item_detailed(item_soup) -> dict | None:
    asin = item_soup.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item_soup.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = extract_brand(title)
    count, pack = extract_count_pack_from_text(title)

    if not count:
        badge = item_soup.select_one(
            "span.a-background-color-platinum, span.a-padding-mini"
        )
        if badge:
            count, pack = extract_count_pack_from_text(badge.get_text())

    # Rating
    rating = ""
    star_icon = item_soup.select_one(
        "i.a-icon-star-small span.a-icon-alt, i.a-icon-star span.a-icon-alt"
    )
    if star_icon:
        rating = star_icon.get_text().split()[0]

    if not rating:
        rating_span = item_soup.select_one("span.a-size-small.a-color-base")
        if rating_span and re.fullmatch(
            r"[\d.]+", rating_span.get_text(strip=True)
        ):
            rating = rating_span.get_text(strip=True)

    # Rating count
    rating_count = ""
    rc_tag = item_soup.select_one("span.a-size-base.s-underline-text")
    if rc_tag:
        rating_count = re.sub(r"[^\d]", "", rc_tag.get_text())

    if not rating_count:
        rc_link = item_soup.select_one("a[aria-label*='ratings']")
        if rc_link:
            m = re.search(
                r"([\d,]+)\s+ratings?", rc_link.get("aria-label", ""), re.I
            )
            if m:
                rating_count = m.group(1).replace(",", "")

    # Prices
    current_price = ""
    p_tag = item_soup.select_one("span.a-price span.a-offscreen")
    if p_tag:
        current_price = p_tag.get_text()

    list_price = ""
    lp_tag = item_soup.select_one(
        "span.a-price.a-text-price span.a-offscreen"
    )
    if lp_tag:
        list_price = lp_tag.get_text()

    total_count = ""
    try:
        total_count = str(int(count) * int(pack))
    except Exception:
        pass

    # Price per count
    ppc = ""
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)
    if cur_val and total_val:
        ppc = f"${(cur_val / total_val):.4f}/count"
    else:
        ppc_tag = item_soup.select_one("span.a-size-base.a-color-secondary")
        if ppc_tag and "/" in ppc_tag.get_text():
            ppc = ppc_tag.get_text()

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": ppc,
    }


async def close_amazon_modals(page) -> bool:
    """
    Try to dismiss Amazon's redirect/region/sign-in overlays.
    Safe to call even if nothing is present.
    """
    selectors = [
        "#redir-modal",                 # the blocking div you see in logs
        "#redir-modal .redir-dismiss",
        ".redir-dismiss",
        "button[aria-label='Close']",
        "button[data-action='a-popover-close']",
    ]

    for sel in selectors:
        try:
            loc = page.locator(sel)
            if await loc.count() > 0:
                await loc.first.click(timeout=3000, force=True)
                await page.wait_for_timeout(500)
                print(f"    Closed modal via selector: {sel}")
                return True
        except PlaywrightTimeoutError:
            continue
        except Exception:
            continue
    return False


# ── SCRAPER ENGINE ──────────────────────────────────────────────────────

async def run_scraper():
    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox"],
        )
        context = await browser.new_context(
            viewport={"width": 1440, "height": 2500},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                "AppleWebKit/537.36 (KHTML, like Gecko) "
                "Chrome/123.0.0.0 Safari/537.36"
            ),
        )
        page = await context.new_page()
        page.set_default_timeout(60000)
        page.set_default_navigation_timeout(90000)

        all_results: list[dict] = []
        seen_asins: set[str] = set()

        for base_url in SEARCH_URLS:
            print(f"Starting Slice: {base_url}")
            try:
                await page.goto(
                    base_url,
                    wait_until="domcontentloaded",
                    timeout=90000,
                )
                await close_amazon_modals(page)

                for page_idx in range(1, MAX_PAGES_PER_SLICE + 1):
                    print(f"  Scraping Page {page_idx}...")

                    # Wait for search results; bail out if nothing loads
                    try:
                        await page.wait_for_selector(
                            "[data-component-type='s-search-result']",
                            timeout=20000,
                        )
                    except PlaywrightTimeoutError:
                        print("    No results loaded; skipping this slice.")
                        break

                    # Scroll to make sure lazy-loaded bits appear
                    for _ in range(3):
                        await page.mouse.wheel(0, 1500)
                        await page.wait_for_timeout(1000)

                    content = await page.content()
                    soup = BeautifulSoup(content, "html.parser")
                    items = soup.select(
                        "[data-component-type='s-search-result']"
                    )

                    for item in items:
                        data = parse_item_detailed(item)
                        if data and data["ASIN"] not in seen_asins:
                            data["Page"] = page_idx
                            all_results.append(data)
                            seen_asins.add(data["ASIN"])

                    # Handle pagination
                    try:
                        await close_amazon_modals(page)
                        next_btn = await page.query_selector(
                            "a.s-pagination-next"
                        )
                        if next_btn:
                            await next_btn.click(timeout=60000)
                            await page.wait_for_timeout(
                                random.randint(3000, 5000)
                            )
                            await close_amazon_modals(page)
                        else:
                            print("    No next button; ending slice.")
                            break
                    except PlaywrightTimeoutError as e:
                        print(f"    Pagination click timed out: {e}")
                        break
                    except Exception as e:
                        print(f"    Unexpected pagination error: {e}")
                        break

            except PlaywrightTimeoutError as e:
                print(f"  Slice navigation timed out: {e}")
                continue
            except Exception as e:
                print(f"  Unexpected error on slice: {e}")
                continue

        if all_results:
            keys = all_results[0].keys()
            with open(OUTPUT_CSV, "w", newline="", encoding="utf-8-sig") as f:
                writer = csv.DictWriter(f, fieldnames=keys)
                writer.writeheader()
                writer.writerows(all_results)
            print(f"Done! Saved {len(all_results)} unique items to {OUTPUT_CSV}")
        else:
            print("No results collected; nothing written.")

        await browser.close()


await run_scraper()
````

````python
# Only extract the unique ASINs on every Amazon page.

import csv
import re
from pathlib import Path
from bs4 import BeautifulSoup


# ── CONFIG ──────────────────────────────────────────────────────────────

HTML_FILES = [
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_one.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_two.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_three.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_four.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_five.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_six.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_Flushable Wipes_seven.html"),
]

OUTPUT_CSV = Path("/content/amazon_flushable_wipes.csv")


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except ValueError:
        return None


def format_price_per_count(value: float) -> str:
    return f"${value:.4f}/count"


def extract_inner_html(path: Path) -> str:
    raw = path.read_text(encoding="utf-8", errors="replace")
    outer = BeautifulSoup(raw, "html.parser")
    cells = outer.select("td.line-content")
    if cells:
        return "\n".join(td.get_text() for td in cells)
    return raw


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        r"(\d[\d,]*)\s+count\s*\(\s*pack\s+of\s+(\d+)\s*\)",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top\s+|resealable\s+|soft\s+|travel\s+)?packs?\s*[,;]?\s*(\d+)\s+(?:wipes?|count|sheets?|cloths?)\s+per\s+pack",
        r"(\d+)\s+(?:wipes?|count|sheets?)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    m = re.search(checks[0], t, re.I)
    if m:
        return m.group(1).replace(",", ""), m.group(2)

    m = re.search(checks[1], t, re.I)
    if m:
        return m.group(1).replace(",", ""), m.group(2)

    m = re.search(checks[2], t, re.I)
    if m:
        return m.group(2), m.group(1)

    m = re.search(checks[3], t, re.I)
    if m:
        per, total = int(m.group(1)), int(m.group(2))
        return str(per), str(total // per if per else 1)

    m = re.search(checks[4], t, re.I)
    if m:
        pack, count, total = int(m.group(1)), int(m.group(2)), int(m.group(3))
        if pack and count:
            return str(count), str(pack)

    m = re.search(checks[5], t, re.I)
    if m:
        return m.group(2), m.group(1)

    pack_match = re.search(r"(\d+)\s+(?:flip-top\s+)?packs?\b", t, re.I)
    per_match = re.search(r"(\d+)\s+(?:wipes?|count|sheets?)\s+per\s+pack", t, re.I)
    total_match = re.search(r"(\d+)\s+total\s+(?:flushable\s+)?wipes?\b|(\d+)\s+total\b", t, re.I)

    if pack_match and per_match:
        return per_match.group(1), pack_match.group(1)

    if pack_match and total_match:
        pack = int(pack_match.group(1))
        total = int(total_match.group(1) or total_match.group(2))
        if pack:
            return str(total // pack), str(pack)

    return "", ""


def extract_count_pack_from_badge(item) -> tuple[str, str]:
    selectors = [
        "span.a-size-base.a-color-base.s-background-color-platinum",
        "span.a-size-base.a-color-base.a-padding-mini",
        "[data-csa-c-asin] span.a-size-base.a-color-base",
    ]

    for selector in selectors:
        for tag in item.select(selector):
            txt = clean(tag.get_text(" ", strip=True))
            count, pack = extract_count_pack_from_text(txt)
            if count and pack:
                return count, pack

    return "", ""


def extract_count_pack(item, title: str) -> tuple[str, str]:
    count, pack = extract_count_pack_from_badge(item)
    if count and pack:
        return count, pack

    count, pack = extract_count_pack_from_text(title)
    if count and pack:
        return count, pack

    return "", ""


def compute_total_count(count: str, pack: str) -> str:
    try:
        return str(int(str(count).replace(",", "")) * int(str(pack).replace(",", "")))
    except:
        return ""


def extract_price_per_count(item, current_price: str, total_count: str) -> str:
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)

    if cur_val and total_val:
        return format_price_per_count(cur_val / total_val)

    all_text = clean(item.get_text(" ", strip=True))
    m = re.search(r"(\$?\d+(?:\.\d+)?)\s*/\s*count", all_text, re.I)
    if m:
        val = m.group(1)
        if not val.startswith("$"):
            val = "$" + val
        return f"{val}/count"

    if cur_val:
        for txt in item.stripped_strings:
            txt = clean(txt)
            val = parse_price_float(txt)
            if txt.startswith("$") and val and val < cur_val:
                return f"{txt}/count"

    return ""


def parse_item(item) -> dict | None:
    asin = item.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""
    if not title:
        return None

    brand = extract_brand(title)
    count, pack = extract_count_pack(item, title)
    total_count = compute_total_count(count, pack)

    rating = ""
    for rs in item.select("span.a-size-small.a-color-base"):
        t = rs.get_text(strip=True)
        if re.fullmatch(r"[\d.]+", t):
            rating = t
            break

    rating_count = ""
    ra = item.select_one("a[aria-label*='ratings']")
    if ra:
        m = re.search(r"([\d,]+)\s+ratings?", ra.get("aria-label", ""), re.I)
        if m:
            rating_count = m.group(1)

    current_price = ""
    for os in item.select("span.a-price span.a-offscreen"):
        txt = clean(os.get_text())
        if txt.startswith("$"):
            current_price = txt
            break

    list_price = ""
    cur_val = parse_price_float(current_price)

    for os in item.select("span.a-price.a-text-price span.a-offscreen"):
        txt = clean(os.get_text())
        val = parse_price_float(txt)
        if txt.startswith("$") and val and cur_val and val > cur_val:
            list_price = txt
            break

    if not list_price:
        for os in item.select("span.a-offscreen"):
            txt = clean(os.get_text())
            if txt.startswith("List:"):
                list_price = txt.replace("List:", "").strip()
                break

    price_per_count = extract_price_per_count(item, current_price, total_count)

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": price_per_count,
    }


def parse_file(html_path: Path) -> list[dict]:
    inner_html = extract_inner_html(html_path)
    soup = BeautifulSoup(inner_html, "html.parser")
    items = soup.select("[data-component-type='s-search-result']")

    records = []
    for item in items:
        rec = parse_item(item)
        if rec:
            records.append(rec)
    return records


# ── MAIN ────────────────────────────────────────────────────────────────

def main():
    all_records = []
    seen_asins = set()

    for html_file in HTML_FILES:
        if not html_file.exists():
            print(f"Warning: file not found -> {html_file}")
            continue

        print(f"Parsing: {html_file.name}")
        records = parse_file(html_file)

        for rec in records:
            asin = rec.get("ASIN", "").strip()
            if asin and asin not in seen_asins:
                seen_asins.add(asin)
                all_records.append(rec)

    fieldnames = [
        "Brand", "ProductTitle", "ASIN", "Count", "Pack", "TotalCount",
        "Rating", "RatingCount", "ListPrice", "CurrentPrice", "PricePerCount",
    ]

    with OUTPUT_CSV.open("w", newline="", encoding="utf-8-sig") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_records)

    print(f"Wrote {len(all_records)} unique rows -> {OUTPUT_CSV}")

    try:
        from google.colab import files
        files.download(str(OUTPUT_CSV))
    except ImportError:
        pass


main()
````

````python
# Only extract baby wipes <$8 on every Amazon page.

import csv
import re
from pathlib import Path
from bs4 import BeautifulSoup


# ── CONFIG ──────────────────────────────────────────────────────────────

HTML_FILES = [
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars one.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars two.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars three.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars four.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars five.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars six.html"),
    Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon.com_ Baby Wipes under 8 dollars seven.html"),
]

OUTPUT_CSV = Path("/content/amazon_baby_wipes_under_8dollars.csv")


# ── HELPERS ─────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except ValueError:
        return None


def format_price_per_count(value: float) -> str:
    return f"${value:.4f}/count"


def extract_inner_html(path: Path) -> str:
    raw = path.read_text(encoding="utf-8", errors="replace")
    outer = BeautifulSoup(raw, "html.parser")
    cells = outer.select("td.line-content")
    if cells:
        return "\n".join(td.get_text() for td in cells)
    return raw


def extract_brand(title: str) -> str:
    if not title:
        return ""
    m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
    return m.group(1).strip() if m else ""


def extract_count_pack_from_text(text: str) -> tuple[str, str]:
    t = clean(text or "")
    if not t:
        return "", ""

    checks = [
        r"(\d[\d,]*)\s+count\s*\(\s*pack\s+of\s+(\d+)\s*\)",
        r"(\d[\d,]*)\s+count\s+pack\s+of\s+(\d+)",
        r"(\d+)\s+(?:flip-top\s+|resealable\s+|soft\s+|travel\s+)?packs?\s*[,;]?\s*(\d+)\s+(?:wipes?|count|sheets?|cloths?)\s+per\s+pack",
        r"(\d+)\s+(?:wipes?|count|sheets?)\s+per\s+pack[^(]*\((\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)\s*,\s*(\d+)\s+total",
        r"(\d+)\s+packs?\s+of\s+(\d+)",
    ]

    m = re.search(checks[0], t, re.I)
    if m:
        return m.group(1).replace(",", ""), m.group(2)

    m = re.search(checks[1], t, re.I)
    if m:
        return m.group(1).replace(",", ""), m.group(2)

    m = re.search(checks[2], t, re.I)
    if m:
        return m.group(2), m.group(1)

    m = re.search(checks[3], t, re.I)
    if m:
        per, total = int(m.group(1)), int(m.group(2))
        return str(per), str(total // per if per else 1)

    m = re.search(checks[4], t, re.I)
    if m:
        pack, count, total = int(m.group(1)), int(m.group(2)), int(m.group(3))
        if pack and count:
            return str(count), str(pack)

    m = re.search(checks[5], t, re.I)
    if m:
        return m.group(2), m.group(1)

    pack_match = re.search(r"(\d+)\s+(?:flip-top\s+)?packs?\b", t, re.I)
    per_match = re.search(r"(\d+)\s+(?:wipes?|count|sheets?)\s+per\s+pack", t, re.I)
    total_match = re.search(r"(\d+)\s+total\s+(?:flushable\s+)?wipes?\b|(\d+)\s+total\b", t, re.I)

    if pack_match and per_match:
        return per_match.group(1), pack_match.group(1)

    if pack_match and total_match:
        pack = int(pack_match.group(1))
        total = int(total_match.group(1) or total_match.group(2))
        if pack:
            return str(total // pack), str(pack)

    return "", ""


def extract_count_pack_from_badge(item) -> tuple[str, str]:
    selectors = [
        "span.a-size-base.a-color-base.s-background-color-platinum",
        "span.a-size-base.a-color-base.a-padding-mini",
        "[data-csa-c-asin] span.a-size-base.a-color-base",
    ]

    for selector in selectors:
        for tag in item.select(selector):
            txt = clean(tag.get_text(" ", strip=True))
            count, pack = extract_count_pack_from_text(txt)
            if count and pack:
                return count, pack

    return "", ""


def extract_count_pack(item, title: str) -> tuple[str, str]:
    count, pack = extract_count_pack_from_badge(item)
    if count and pack:
        return count, pack

    count, pack = extract_count_pack_from_text(title)
    if count and pack:
        return count, pack

    return "", ""


def compute_total_count(count: str, pack: str) -> str:
    try:
        return str(int(str(count).replace(",", "")) * int(str(pack).replace(",", "")))
    except:
        return ""


def extract_price_per_count(item, current_price: str, total_count: str) -> str:
    cur_val = parse_price_float(current_price)
    total_val = parse_price_float(total_count)

    if cur_val and total_val:
        return format_price_per_count(cur_val / total_val)

    all_text = clean(item.get_text(" ", strip=True))
    m = re.search(r"(\$?\d+(?:\.\d+)?)\s*/\s*count", all_text, re.I)
    if m:
        val = m.group(1)
        if not val.startswith("$"):
            val = "$" + val
        return f"{val}/count"

    if cur_val:
        for txt in item.stripped_strings:
            txt = clean(txt)
            val = parse_price_float(txt)
            if txt.startswith("$") and val and val < cur_val:
                return f"{txt}/count"

    return ""


def parse_item(item) -> dict | None:
    asin = item.get("data-asin", "").strip()
    if not asin:
        return None

    h2 = item.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""
    if not title:
        return None

    brand = extract_brand(title)
    count, pack = extract_count_pack(item, title)
    total_count = compute_total_count(count, pack)

    rating = ""
    for rs in item.select("span.a-size-small.a-color-base"):
        t = rs.get_text(strip=True)
        if re.fullmatch(r"[\d.]+", t):
            rating = t
            break

    rating_count = ""
    ra = item.select_one("a[aria-label*='ratings']")
    if ra:
        m = re.search(r"([\d,]+)\s+ratings?", ra.get("aria-label", ""), re.I)
        if m:
            rating_count = m.group(1)

    current_price = ""
    for os in item.select("span.a-price span.a-offscreen"):
        txt = clean(os.get_text())
        if txt.startswith("$"):
            current_price = txt
            break

    list_price = ""
    cur_val = parse_price_float(current_price)

    for os in item.select("span.a-price.a-text-price span.a-offscreen"):
        txt = clean(os.get_text())
        val = parse_price_float(txt)
        if txt.startswith("$") and val and cur_val and val > cur_val:
            list_price = txt
            break

    if not list_price:
        for os in item.select("span.a-offscreen"):
            txt = clean(os.get_text())
            if txt.startswith("List:"):
                list_price = txt.replace("List:", "").strip()
                break

    price_per_count = extract_price_per_count(item, current_price, total_count)

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "TotalCount": total_count,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": price_per_count,
    }


def parse_file(html_path: Path) -> list[dict]:
    inner_html = extract_inner_html(html_path)
    soup = BeautifulSoup(inner_html, "html.parser")
    items = soup.select("[data-component-type='s-search-result']")

    records = []
    for item in items:
        rec = parse_item(item)
        if rec:
            records.append(rec)
    return records


# ── MAIN ────────────────────────────────────────────────────────────────

def main():
    all_records = []
    seen_asins = set()

    for html_file in HTML_FILES:
        if not html_file.exists():
            print(f"Warning: file not found -> {html_file}")
            continue

        print(f"Parsing: {html_file.name}")
        records = parse_file(html_file)

        for rec in records:
            asin = rec.get("ASIN", "").strip()
            if asin and asin not in seen_asins:
                seen_asins.add(asin)
                all_records.append(rec)

    fieldnames = [
        "Brand", "ProductTitle", "ASIN", "Count", "Pack", "TotalCount",
        "Rating", "RatingCount", "ListPrice", "CurrentPrice", "PricePerCount",
    ]

    with OUTPUT_CSV.open("w", newline="", encoding="utf-8-sig") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_records)

    print(f"Wrote {len(all_records)} unique rows -> {OUTPUT_CSV}")

    try:
        from google.colab import files
        files.download(str(OUTPUT_CSV))
    except ImportError:
        pass


main()
````

````python
import csv
import re
from pathlib import Path
from bs4 import BeautifulSoup


# ── config ────────────────────────────────────────────────────────────────────

INPUT_HTML = Path("/content/drive/My Drive/Personal Care Project/Web Scraping_Wipes/Amazon_Flushable_Wipes_I.html")
OUTPUT_CSV = Path("/content/amazon_flushable_wipes.csv")


# ── helpers ───────────────────────────────────────────────────────────────────

def clean(text: str) -> str:
    return re.sub(r"\s+", " ", (text or "")).strip()


def parse_price_float(s: str):
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except ValueError:
        return None


# ── inner HTML extraction (Chrome source-viewer wrapper) ─────────────────────

def extract_inner_html(path: Path) -> str:
    raw = path.read_text(encoding="utf-8", errors="replace")
    outer = BeautifulSoup(raw, "html.parser")
    cells = outer.select("td.line-content")
    if cells:
        return "\n".join(td.get_text() for td in cells)
    return raw


# ── field extractors ──────────────────────────────────────────────────────────

def extract_count_pack(title: str) -> tuple[str, str]:
    t = title

    m = re.search(r"(\d[\d,]*)\s+count\s*\((\d+)\s+packs?\s+of\s+(\d+)\)", t, re.I)
    if m:
        return m.group(3), m.group(2)

    m = re.search(
        r"(\d+)\s+(?:\w[\w\-]*\s+)?packs?\s*[,;]\s*(\d+)\s+"
        r"(?:wipes?|count|sheets?|cloths?)\s+per\s+pack",
        t, re.I
    )
    if m:
        return m.group(2), m.group(1)

    m = re.search(
        r"(\d+)\s+(?:wipes?|count|sheets?)\s+per\s+pack[^(]*\((\d+)\s+total",
        t, re.I
    )
    if m:
        per, total = int(m.group(1)), int(m.group(2))
        return str(per), str(total // per if per else 1)

    m = re.search(
        r"(\d[\d,]*)\s*(?:count|wipes?)\s*\(\s*(?:pack\s+of|pack)\s*(\d+)\s*\)",
        t, re.I
    )
    if m:
        total, packs = int(m.group(1).replace(",", "")), int(m.group(2))
        return str(total // packs if packs else total), str(packs)

    m = re.search(r"(\d[\d,]*)\s+wipes?\s*\((\d+)\s+packs?\s+of\s+(\d+)\)", t, re.I)
    if m:
        return m.group(3), m.group(2)

    m = re.search(r"(\d+)\s+resealable\s+packets?\s*=\s*(\d+)\s+total", t, re.I)
    if m:
        packs, total = int(m.group(1)), int(m.group(2))
        return str(total // packs if packs else total), str(packs)

    m = re.search(r"(\d[\d,]*)\s+count\b", t, re.I)
    if m:
        return m.group(1).replace(",", ""), "1"

    m = re.search(r"(\d[\d,]*)\s+(?:total\s+)?wipes?\b", t, re.I)
    if m:
        return m.group(1).replace(",", ""), "1"

    return "", ""


def parse_item(item) -> dict:
    asin = item.get("data-asin", "").strip()

    h2 = item.select_one("h2 span")
    title = clean(h2.get_text()) if h2 else ""

    brand = ""
    if title:
        m = re.match(r"^([A-Z][A-Za-z&\-]+(?:\s+[A-Z][A-Za-z&\-]+){0,2})", title)
        if m:
            brand = m.group(1).strip()

    count, pack = extract_count_pack(title)

    rating = ""
    for rs in item.select("span.a-size-small.a-color-base"):
        t = rs.get_text(strip=True)
        if re.fullmatch(r"[\d.]+", t):
            rating = t
            break

    rating_count = ""
    ra = item.select_one("a[aria-label*='ratings']")
    if ra:
        m = re.search(r"([\d,]+)\s+ratings?", ra.get("aria-label", ""), re.I)
        if m:
            rating_count = m.group(1)

    current_price = ""
    cp = item.select_one("span.a-price[data-a-color='base'] span.a-offscreen")
    if cp:
        current_price = clean(cp.get_text())

    price_per_count = ""
    cur_val = parse_price_float(current_price)

    for txt in item.stripped_strings:
        s = clean(txt)
        if "/count" in s.lower():
            m = re.search(r"\$?\d+(?:\.\d+)?\s*/\s*count", s, re.I)
            if m:
                val = m.group(0).replace(" ", "")
                if not val.startswith("$"):
                    val = "$" + val
                price_per_count = val.lower().replace("/Count", "/count")
                break

    if not price_per_count:
        for tag in item.select("span.a-price.a-text-price"):
            if tag.get("data-a-strike"):
                continue
            os = tag.select_one("span.a-offscreen")
            if not os:
                continue
            val_str = clean(os.get_text())
            val = parse_price_float(val_str)
            if cur_val and val and val < cur_val:
                price_per_count = val_str + "/count"
                break

    list_price = ""
    for tag in item.select("span.a-price.a-text-price"):
        if tag.get("data-a-strike") == "true":
            os = tag.select_one("span.a-offscreen")
            if os:
                list_price = clean(os.get_text())
                break

    if not list_price:
        for os in item.select("span.a-offscreen"):
            t = clean(os.get_text())
            if t.startswith("List:"):
                list_price = t.replace("List:", "").strip()
                break

    return {
        "Brand": brand,
        "ProductTitle": title,
        "ASIN": asin,
        "Count": count,
        "Pack": pack,
        "Rating": rating,
        "RatingCount": rating_count,
        "ListPrice": list_price,
        "CurrentPrice": current_price,
        "PricePerCount": price_per_count,
    }


# ── main parsing ──────────────────────────────────────────────────────────────

FIELDNAMES = [
    "Brand", "ProductTitle", "ASIN", "Count", "Pack",
    "Rating", "RatingCount", "ListPrice", "CurrentPrice", "PricePerCount",
]


def parse_file(html_path: Path) -> list[dict]:
    inner_html = extract_inner_html(html_path)
    soup = BeautifulSoup(inner_html, "html.parser")
    items = soup.select("[data-component-type='s-search-result']")

    seen = set()
    records = []

    for item in items:
        asin = item.get("data-asin", "").strip()
        if not asin or asin in seen:
            continue
        seen.add(asin)
        records.append(parse_item(item))

    return records


def write_csv(records: list[dict], out_path: Path) -> None:
    with out_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=FIELDNAMES)
        writer.writeheader()
        writer.writerows(records)
    print(f"Wrote {len(records)} rows -> {out_path}")


def main():
    if not INPUT_HTML.exists():
        raise FileNotFoundError(
            f"Input file not found: {INPUT_HTML}\n"
            f"Upload the HTML into Colab first, then update INPUT_HTML if needed."
        )

    records = parse_file(INPUT_HTML)
    write_csv(records, OUTPUT_CSV)

    try:
        from google.colab import files
        files.download(str(OUTPUT_CSV))
    except ImportError:
        print("Not running in Google Colab, so automatic download was skipped.")


if __name__ == "__main__":
    main()
````

````python
"""
generate_wipes_report.py
------------------------
Block 2: reads the parsed CSV from parse_amazon_wipes.py and produces
an enriched "Amazon Flushable Wipes" report CSV with additional computed
columns:

    TotalCount       = Count × Pack  (total wipes in the listing)
    CalcPricePerCount= CurrentPrice ÷ TotalCount  (recalculated from prices)
    Discount         = % markdown off ListPrice  (blank if no list price)
    ValueScore       = Rating ÷ CalcPricePerCount  (higher = better value)

Output columns (in order):
    Brand, ProductTitle, ASIN,
    Count, Pack, TotalCount,
    Rating, RatingCount,
    ListPrice, CurrentPrice,
    PricePerCount, CalcPricePerCount,
    Discount, ValueScore

Usage
-----
    python generate_wipes_report.py                          # default I/O
    python generate_wipes_report.py parsed.csv               # custom input
    python generate_wipes_report.py parsed.csv report.csv    # custom I/O

    # Run both steps end-to-end:
    python parse_amazon_wipes.py && python generate_wipes_report.py
"""

import csv
import re
import sys
from pathlib import Path


# ── helpers ───────────────────────────────────────────────────────────────────

def to_float(s: str):
    """Strip currency symbols / commas and return float, or None."""
    try:
        return float(re.sub(r"[^\d.]", "", s or ""))
    except ValueError:
        return None


def fmt_dollars(val) -> str:
    """Format a float as '$X.XX', or '' if None."""
    return f"${val:.2f}" if val is not None else ""


def fmt_pct(val) -> str:
    """Format a float as 'X.X%', or '' if None."""
    return f"{val:.1f}%" if val is not None else ""


def fmt_score(val) -> str:
    """Format value score to 1 decimal place, or '' if None."""
    return f"{val:.1f}" if val is not None else ""


# ── enrichment ────────────────────────────────────────────────────────────────

def enrich(row: dict) -> dict:
    count     = to_float(row.get("Count"))
    pack      = to_float(row.get("Pack")) or 1
    cur_price = to_float(row.get("CurrentPrice"))
    lst_price = to_float(row.get("ListPrice"))
    rating    = to_float(row.get("Rating"))

    # TotalCount
    total_count = int(count * pack) if count else None

    # CalcPricePerCount  (always recalculate from price / total for consistency)
    calc_ppc = None
    if cur_price and total_count:
        calc_ppc = round(cur_price / total_count, 4)

    # Discount %
    discount = None
    if lst_price and cur_price and lst_price > cur_price:
        discount = round((lst_price - cur_price) / lst_price * 100, 1)

    # ValueScore = Rating / CalcPricePerCount  (higher = better value per dollar)
    value_score = None
    if rating and calc_ppc and calc_ppc > 0:
        value_score = round(rating / calc_ppc, 1)

    return {
        "Brand":             row.get("Brand", ""),
        "ProductTitle":      row.get("ProductTitle", ""),
        "ASIN":              row.get("ASIN", ""),
        "Count":             row.get("Count", ""),
        "Pack":              row.get("Pack", ""),
        "TotalCount":        total_count if total_count is not None else "",
        "Rating":            row.get("Rating", ""),
        "RatingCount":       row.get("RatingCount", ""),
        "ListPrice":         row.get("ListPrice", ""),
        "CurrentPrice":      row.get("CurrentPrice", ""),
        "PricePerCount":     row.get("PricePerCount", ""),
        "CalcPricePerCount": fmt_dollars(calc_ppc),
        "Discount":          fmt_pct(discount),
        "ValueScore":        fmt_score(value_score),
    }


# ── I/O ───────────────────────────────────────────────────────────────────────

FIELDNAMES = [
    "Brand", "ProductTitle", "ASIN",
    "Count", "Pack", "TotalCount",
    "Rating", "RatingCount",
    "ListPrice", "CurrentPrice",
    "PricePerCount", "CalcPricePerCount",
    "Discount", "ValueScore",
]


def generate_report(in_path: Path, out_path: Path) -> None:
    with in_path.open(encoding="utf-8") as f:
        rows = list(csv.DictReader(f))

    enriched = [enrich(r) for r in rows]

    # Sort by CalcPricePerCount ascending (best value first), blanks last
    def sort_key(r):
        v = to_float(r["CalcPricePerCount"])
        return (v is None, v or 0)

    enriched.sort(key=sort_key)

    with out_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=FIELDNAMES)
        writer.writeheader()
        writer.writerows(enriched)

    print(f"✅  Enriched report → {out_path}  ({len(enriched)} products)")
    print(f"\nTop 5 by price-per-count:")
    for r in enriched[:5]:
        print(f"  {r['ASIN']}  {r['Brand'][:22]:<22}  "
              f"Total={str(r['TotalCount']):<5}  "
              f"CurPrice={r['CurrentPrice']:<8}  "
              f"$/count={r['CalcPricePerCount']:<8}  "
              f"Rating={r['Rating']}  Score={r['ValueScore']}")


# ── entry point ───────────────────────────────────────────────────────────────

if __name__ == "__main__":
    args = sys.argv[1:]
    in_path  = Path(args[0]) if len(args) >= 1 else Path("amazon_flushable_wipes.csv")
    out_path = Path(args[1]) if len(args) >= 2 else Path("amazon_flushable_wipes_report.csv")

    if not in_path.exists():
        print(f"❌  Input not found: {in_path}")
        print("    Run parse_amazon_wipes.py first to generate it.")
        sys.exit(1)

    generate_report(in_path, out_path)
````

````python
import pandas as pd
from openai import OpenAI
from io import StringIO
from google.colab import files
import os

from google.colab import userdata
a = userdata.get('api_key')

# Initialize client with your API key
# Ensure your API key is correctly set up.
# REPLACE "YOUR_VALID_OPENAI_API_KEY_HERE" WITH YOUR ACTUAL KEY
# Edition: May need to update the api_key
client = OpenAI(api_key= a)

# --- 1. Load Full Keyword Data and Clean ---
# Edition: Update the file path
file_path = "/content/drive/My Drive/Personal Care Project/Google wipes Keyword Stats 2026-02-11 at 09_12_45.csv"
print(f"Loading data from: {file_path}")
df_main = pd.read_csv(file_path, skiprows=2)

# Drop any entirely empty columns that might result from skiprows or file structure
df_main = df_main.dropna(axis=1, how='all')

# Ensure essential columns exist and clean them
if 'Keyword' not in df_main.columns or 'Avg. monthly searches' not in df_main.columns:
    raise ValueError("DataFrame must contain 'Keyword' and 'Avg. monthly searches' columns.")
df_main = df_main[['Keyword', 'Avg. monthly searches']].dropna(subset=['Keyword', 'Avg. monthly searches'])

# Convert 'Avg. monthly searches' to numeric, handling potential errors and filling NaNs
df_main['Avg. monthly searches'] = pd.to_numeric(df_main['Avg. monthly searches'], errors='coerce')
df_main.dropna(subset=['Avg. monthly searches'], inplace=True)
df_main['Avg. monthly searches'] = df_main['Avg. monthly searches'].astype(int)

print("Cleaned DataFrame Head:")
print(df_main.head())
print(f"Total keywords after cleaning: {len(df_main)}")

# --- 2. Define Batch Classification Function ---
# update prompt consumer product contents
def classify_batch(batch_df: pd.DataFrame, batch_id: int) -> pd.DataFrame:
    """
    Classifies a batch of keywords using the OpenAI API.

    Args:
        batch_df: DataFrame containing 'Keyword' and 'Avg. monthly searches' columns.
        batch_id: Identifier for the current batch.

    Returns:
        DataFrame with classified results or an empty DataFrame if an error occurs.
    """
    keywords_csv_content = batch_df.to_csv(index=False)

    prompt = f"""
    You are an expert in wipes, flushable wipes, XL wipes, Bamboo wipes consumer insights.
    Classify *all* of the following keywords into one or more of these categories.
    Do not skip any keywords, regardless of their search volume.
    If a keyword belongs to multiple categories, list them separated by a pipe symbol "|".


    **Predefined Categories:**
    - Brand
    - Claim
    - Format
    - Ingredient
    - Function/Benefit
    - Variants/Types
    - Scent
    - User

    **Output Format:**
    For each keyword, return a CSV row with exactly three columns, including a header row:
    Keyword,Category,Avg. monthly searches

    Ensure the Avg. monthly searches value is exactly as provided in the input.


    **Keywords to Classify (CSV format, include header row):**
    Keyword,Avg. monthly searches
    {keywords_csv_content}
    """

    print(f"Sending batch {batch_id} with {len(batch_df)} keywords to OpenAI...")
    try:
        response = client.chat.completions.create(
            model="gpt-4o", # Using gpt-4o as gpt-5 is a non-standard model name.
            temperature=1, # Explicitly setting temperature to 1 as 0 is not supported for some models.
            messages=[{"role": "user", "content": prompt}],
            response_format={"type": "text"} # Explicitly request text output
        )
        output_text = response.choices[0].message.content

        # Attempt to parse as CSV, robustly handling potential variations in model output
        try:
            # Try parsing assuming the model provides a header row
            classified_df = pd.read_csv(StringIO(output_text.strip()))
            # Validate columns
            if classified_df.columns.tolist() != ['Keyword', 'Category', 'Avg. monthly searches']:
                 # If columns are not as expected, try parsing without header and assign names
                print(f"Warning: Batch {batch_id} - CSV header mismatch. Trying parsing without header.")
                classified_df = pd.read_csv(StringIO(output_text.strip()), header=None, names=['Keyword', 'Category', 'Avg. monthly searches'])
                if classified_df.isnull().any().any(): # Check if this results in NaNs for columns
                    print(f"Warning: Batch {batch_id} - Parsing without header resulted in NaNs. Fallback to basic comma split.")
                    raise ValueError("Failed to parse cleanly with pandas.") # Trigger fallback
            return classified_df
        except (pd.errors.EmptyDataError, ValueError, pd.errors.ParserError) as csv_err:
            print(f"Error parsing CSV output for batch {batch_id} with pandas: {csv_err}")
            print(f"Attempting manual CSV parsing. Output text (first 500 chars): \n{output_text[:500]}...")
            # Manual parsing fallback
            lines = output_text.strip().split('\n')
            data_rows = []
            # Skip header if it exists
            if lines and lines[0].lower().startswith('keyword,category,avg. monthly searches'):
                lines = lines[1:]
            for line in lines:
                parts = line.split(',')
                if len(parts) == 3:
                    # Basic cleanup of parts
                    data_rows.append([p.strip() for p in parts])
                else:
                    print(f"Warning: Batch {batch_id} - Skipping malformed line during manual parse: {line}")
            if data_rows:
                return pd.DataFrame(data_rows, columns=['Keyword', 'Category', 'Avg. monthly searches'])
            else:
                print(f"Error: Batch {batch_id} - Manual parsing also failed to extract data.")
                return pd.DataFrame(columns=['Keyword', 'Category', 'Avg. monthly searches']) # Return empty DF

    except Exception as e:
        print(f"Error calling OpenAI API for batch {batch_id}: {e}")
        return pd.DataFrame(columns=['Keyword', 'Category', 'Avg. monthly searches'])

# --- 3. Process Keywords in Batches ---
batch_size = 500 # User requested batch size was 300-500.
classified_results = []
output_dir = "/content/classified_batches/"
os.makedirs(output_dir, exist_ok=True)
downloaded_files = []

num_batches = (len(df_main) + batch_size - 1) // batch_size
print(f"Processing {len(df_main)} keywords in {num_batches} batches of size {batch_size}...")

for i in range(num_batches):
    start_idx = i * batch_size
    end_idx = min((i + 1) * batch_size, len(df_main))
    batch_df = df_main.iloc[start_idx:end_idx]

    if not batch_df.empty:
        classified_batch_df = classify_batch(batch_df, i + 1)
        if not classified_batch_df.empty:
            # Ensure 'Avg. monthly searches' is numeric before saving and aggregating
            classified_batch_df['Avg. monthly searches'] = pd.to_numeric(
                classified_batch_df['Avg. monthly searches'], errors='coerce'
            ).fillna(0).astype(int)

            batch_filename = os.path.join(output_dir, f"classified_keywords_batch_{i+1}.csv")
            classified_batch_df.to_csv(batch_filename, index=False)
            classified_results.append(classified_batch_df)
            downloaded_files.append(batch_filename)
            print(f"Batch {i+1} classified and saved to {batch_filename}")

            # Download each batch CSV to the user's local machine
            try:
                files.download(batch_filename)
                print(f"Batch {i+1} downloaded.")
            except Exception as e:
                print(f"Could not download batch {i+1} (may be due to Colab environment or browser settings): {e}")
        else:
            print(f"Batch {i+1} returned empty classified results. Skipping save/download.")
    else:
        print(f"Skipping empty batch {i+1}.")

print("\n--- All batches processed ---")

# --- 4. Aggregate and Summarize All Results ---
if classified_results:
    all_classified_df = pd.concat(classified_results, ignore_index=True)

    # Ensure 'Avg. monthly searches' is numeric and handle potential multi-category entries
    all_classified_df['Avg. monthly searches'] = pd.to_numeric(
        all_classified_df['Avg. monthly searches'], errors='coerce'
    ).fillna(0).astype(int)

    # Explode categories for proper aggregation if a keyword has multiple categories
    # Convert Category to string first to handle any non-string types, then split
    all_classified_df['Category'] = all_classified_df['Category'].astype(str).str.split('|')
    exploded_df = all_classified_df.explode('Category')
    exploded_df['Category'] = exploded_df['Category'].str.strip() # Clean whitespace from categories

    # --- NEW: Strict Category Remapping for Consistency ---
    # Define the exact list of desired top-level categories
    desired_top_level_categories = ['Brand', 'Claim', 'Format', 'Ingredient', 'Function/Benefit', 'Variants/Types', 'Scent', 'User']

    # Remap any category not in the desired list to 'Other'
    exploded_df.loc[~exploded_df['Category'].isin(desired_top_level_categories), 'Category'] = 'Other'
    # --- END NEW ---

    aggregated_summary = exploded_df.groupby('Category').agg(
        Total_Monthly_Searches=('Avg. monthly searches', 'sum'),
        Keyword_Count=('Keyword', 'nunique') # Count unique keywords per category
    ).reset_index().sort_values(by='Total_Monthly_Searches', ascending=False)

    print("\n--- Final Aggregated Summary by Category ---")
    print(aggregated_summary)

    summary_filename = "/content/final_classified_summary.csv"
    aggregated_summary.to_csv(summary_filename, index=False)
    print(f"\nFinal summary saved to {summary_filename}")
    try:
        files.download(summary_filename)
        print("Final summary downloaded.")
    except Exception as e:
        print(f"Could not download final summary (may be due to Colab environment or browser settings): {e}")
else:
    print("No classified results to aggregate.")
````

## Define `group_keyword_concept` function and aggregate high-volume keywords

### Subtask:
Define a function to group keywords into concepts, apply it to the exploded DataFrame, filter for high-volume keywords, and then aggregate these into `aggregated_high_volume_concept`.


**Reasoning**:
The subtask requires defining a function `group_keyword_concept` to categorize keywords based on their 'Category' and 'Keyword' values. This function will use specific patterns to create a 'Grouped_Concept', and if no pattern matches, it will return the original 'Keyword'. This is the first step towards creating the `aggregated_high_volume_concept` DataFrame.



````python
import pandas as pd
import re
from google.colab import files

def group_keyword_concept(row):
    category = row['Category']
    keyword = str(row['Keyword']).lower()

    # Default to original keyword if no specific grouping applies
    grouped_concept = row['Keyword']


    if category == 'Brand':
        # Private Label Brands (high priority)
        private_label_brands_pattern = r'walmart|365|equate|parent\s?s choice|target|up & up|cvs|cvs health|cvs natural baby wipes|cvs pure water baby wipes|walgreens|costco|kirkland signature|kirkland|sam|sam\s? club|member\s?s mark|kroger|comforts|publix|aldi|little journey|lidl|lupilu|meijer|h-e-b|albertsons|safeway|signature care|dollar general|dg baby|family dollar|kidgets|family wellness|bj\s?s wholesale|berkley jensen|shoppers drug mart|life brand superstore|loblaws|tesco|fred & flo|sainsbury\s?s|little ones|asda|little angels|morrisons|nutmeg'
        if re.search(private_label_brands_pattern, keyword):
            grouped_concept = 'Private Label'
        # Amazon Private Label Brands (high priority)
        elif re.search(r'amazon|mama bear|amazon elements|Amazon Elements', keyword):
            grouped_concept = 'Amazon Private Label'
        # Other common wipe brands
        elif re.search(r'huggies', keyword): grouped_concept = 'Huggies'
        elif re.search(r'honest', keyword): grouped_concept = 'Honest'
        elif re.search(r'charmin', keyword): grouped_concept = 'Charmin'
        elif re.search(r'kandoo', keyword): grouped_concept = 'Kandoo'
        elif re.search(r'cottonelle', keyword): grouped_concept = 'Cottonelle'
        elif re.search(r'pampers', keyword): grouped_concept = 'Pampers'
        elif re.search(r'waterwipes', keyword): grouped_concept = 'WaterWipes'
        elif re.search(r'wet ones', keyword): grouped_concept = 'Wet Ones'
        elif re.search(r'seventh generation', keyword): grouped_concept = 'Seventh Generation'
        elif re.search(r'dyper', keyword): grouped_concept = 'Dyper'
        elif re.search(r'nice \'n clean|nice n clean|nice and clean', keyword): grouped_concept = 'Nice \'N Clean'
        elif re.search(r'boogie wipes', keyword): grouped_concept = 'Boogie Wipes'
        elif re.search(r'duke cannon', keyword): grouped_concept = 'Duke Cannon'
        elif re.search(r'goodwipes', keyword): grouped_concept = 'Goodwipes'
        elif re.search(r'action wipes', keyword): grouped_concept = 'Action Wipes'
        elif re.search(r'stall mates', keyword): grouped_concept = 'Stall Mates'
        elif re.search(r'dude wipes', keyword): grouped_concept = 'Dude Wipes'
        elif re.search(r'waterwipe|waterwipes', keyword): grouped_concept = 'Dude Wipes'
        elif re.search(r'curash', keyword): grouped_concept = 'Curash'
        elif re.search(r'bumboosa', keyword): grouped_concept = 'Bumboosa'
        elif re.search(r'neosporin', keyword): grouped_concept = 'Neosporin'
        elif re.search(r'kimberly-clark', keyword): grouped_concept = 'Kimberly-Clark'
        elif re.search(r'the cheeky panda', keyword): grouped_concept = 'The Cheeky Panda'

    elif category == 'Claim':
        if re.search(r'flushable|dispersible|flush', keyword): grouped_concept = 'Flushable Claim'
        elif re.search(r'biodegradable|compostable', keyword): grouped_concept = 'Biodegradable/Compostable Claim'
        elif re.search(r'sensitive|hypoallergenic', keyword): grouped_concept = 'Sensitive/Hypoallergenic Claim'
        elif re.search(r'unscented|fragrance free|fragrance-free|without fragrance|scent free|scent-free', keyword): grouped_concept = 'Unscented/Fragrance Free Claim'
        elif re.search(r'natural|organic', keyword): grouped_concept = 'Natural/Organic Claim'
        elif re.search(r'dermatologist tested', keyword): grouped_concept = 'Dermatologist Tested Claim'
        elif re.search(r'alcohol free', keyword): grouped_concept = 'Alcohol-Free Claim'
        elif re.search(r'paraben free', keyword): grouped_concept = 'Paraben-Free Claim'
        elif re.search(r'best|top', keyword): grouped_concept = 'Best/Top Claim'
        elif re.search(r'eco-friendly|eco', keyword): grouped_concept = 'Eco-Friendly Claim'
        elif re.search(r'plant-based|plant', keyword): grouped_concept = 'Plant-based Claim'
        elif re.search(r'durable|strong', keyword): grouped_concept = 'Durable/Strong Claim'
        elif re.search(r'gentle', keyword): grouped_concept = 'Gentle Claim'
        elif re.search(r'plant based|plant-based', keyword): grouped_concept = 'Plant-Based Claim'
        elif re.search(r'pure', keyword): grouped_concept = 'Pure Claim'
        elif re.search(r'disposable', keyword): grouped_concept = 'Disposable Claim'
        elif re.search(r'best|top', keyword): grouped_concept = 'Best/Top Claim'
        elif re.search(r'ph balanced', keyword): grouped_concept = 'pH Balanced Claim'
        elif re.search(r'travel friendly', keyword): grouped_concept = 'Travel Friendly Claim'
        elif re.search(r'bulk', keyword): grouped_concept = 'Bulk Purchase Claim'



    elif category == 'Ingredient':
        if re.search(r'water', keyword): grouped_concept = 'Water'
        elif re.search(r'aloe vera|aloe', keyword): grouped_concept = 'Aloe Vera'
        elif re.search(r'bamboo', keyword): grouped_concept = 'Bamboo'
        elif re.search(r'cotton', keyword): grouped_concept = 'Cotton'
        elif re.search(r'shea butter', keyword): grouped_concept = 'Shea Butter'
        elif re.search(r'vitamin e', keyword): grouped_concept = 'Vitamin E'
        elif re.search(r'glycerin', keyword): grouped_concept = 'Glycerin'
        elif re.search(r'chamomile', keyword): grouped_concept = 'Chamomile'
        elif re.search(r'plant fiber|plant-fiber', keyword): grouped_concept = 'Plant Fiber'
        elif re.search(r'essential oil', keyword): grouped_concept = 'Essential Oil'

    elif category == 'Function/Benefit':
        if re.search(r'cleaning|cleanse', keyword): grouped_concept = 'Cleaning/Cleansing Benefit'
        elif re.search(r'personal hygiene|feminine hygiene', keyword): grouped_concept = 'Personal/Feminine Hygiene Benefit'
        elif re.search(r'toilet|after toilet', keyword): grouped_concept = 'Toilet Use Benefit'
        elif re.search(r'freshening|refreshing', keyword): grouped_concept = 'Freshening/Refreshing Benefit'
        elif re.search(r'moisturizing|hydrating', keyword): grouped_concept = 'Moisturizing/Hydrating Benefit'
        elif re.search(r'makeup remover', keyword): grouped_concept = 'Makeup Remover Benefit'
        elif re.search(r'antiseptic|antibacterial', keyword): grouped_concept = 'Antiseptic/Antibacterial Benefit'
        elif re.search(r'disinfecting', keyword): grouped_concept = 'Disinfecting Benefit'
        elif re.search(r'odor control', keyword): grouped_concept = 'Odor Control Benefit'
        elif re.search(r'travel', keyword): grouped_concept = 'Travel Benefit'
        elif re.search(r'outdoor', keyword): grouped_concept = 'Outdoor Use Benefit'
        elif re.search(r'adult care', keyword): grouped_concept = 'Adult Care Benefit'
        elif re.search(r'incontinence', keyword): grouped_concept = 'Incontinence Care Benefit'
        elif re.search(r'hemorrhoid|hemorrhoidal', keyword): grouped_concept = 'Hemorrhoid Care Benefit'
        elif re.search(r'first aid', keyword): grouped_concept = 'First Aid Benefit'

    elif category == 'Variants/Types':
        if re.search(r'baby wipes|infant wipes', keyword): grouped_concept = 'Baby Wipes'
        elif re.search(r'face wipes|facial wipes', keyword): grouped_concept = 'Face Wipes'
        elif re.search(r'body wipes', keyword): grouped_concept = 'Body Wipes'
        elif re.search(r'toilet wipes', keyword): grouped_concept = 'Toilet Wipes'
        elif re.search(r'feminine wipes|intimate wipes', keyword): grouped_concept = 'Feminine/Intimate Wipes'
        elif re.search(r'adult wipes', keyword): grouped_concept = 'Adult Wipes'
        elif re.search(r'camping wipes|travel wipes', keyword): grouped_concept = 'Camping/Travel Wipes'
        elif re.search(r'gym wipes|sport wipes', keyword): grouped_concept = 'Gym/Sport Wipes'
        elif re.search(r'flushable wipes', keyword): grouped_concept = 'Flushable Wipes'
        elif re.search(r'wet wipes', keyword): grouped_concept = 'Wet Wipes'
        elif re.search(r'dry wipes', keyword): grouped_concept = 'Dry Wipes'
        elif re.search(r'medicated wipes', keyword): grouped_concept = 'Medicated Wipes'
        elif re.search(r'anti-itch wipes', keyword): grouped_concept = 'Anti-Itch Wipes'
        elif re.search(r'dog wipes|pet wipes', keyword): grouped_concept = 'Pet Wipes'
        elif re.search(r'hand wipes', keyword): grouped_concept = 'Hand Wipes'
        elif re.search(r'antibacterial wipes', keyword): grouped_concept = 'Antibacterial Wipes'
        elif re.search(r'xxl|xl wipes', keyword): grouped_concept = 'XL Wipes'
        elif re.search(r'personal wipes', keyword): grouped_concept = 'Personal Wipes'

    elif category == 'Scent':
        if re.search(r'unscented|fragrance free', keyword): grouped_concept = 'Unscented'
        elif re.search(r'fresh scent', keyword): grouped_concept = 'Fresh Scent'
        elif re.search(r'coconut scent', keyword): grouped_concept = 'Coconut Scent'
        elif re.search(r'lavender scent', keyword): grouped_concept = 'Lavender Scent'
        elif re.search(r'citrus scent', keyword): grouped_concept = 'Citrus Scent'

    elif category == 'User':
        if re.search(r'baby|infant', keyword): grouped_concept = 'User: Baby'
        elif re.search(r'kids|children|child', keyword): grouped_concept = 'User: Kids/Children'
        elif re.search(r'adult|adults', keyword): grouped_concept = 'User: Adult'
        elif re.search(r'men', keyword): grouped_concept = 'User: Men'
        elif re.search(r'women|feminine', keyword): grouped_concept = 'User: Women/Feminine'
        elif re.search(r'elderly|senior', keyword): grouped_concept = 'User: Elderly/Senior'
        elif re.search(r'patient|incontinence', keyword): grouped_concept = 'User: Patients/Incontinence'
        elif re.search(r'diaper rash', keyword): grouped_concept = 'User: Diaper Rash Sufferers'
        elif re.search(r'traveler|travel', keyword): grouped_concept = 'User: Traveler'
        elif re.search(r'camper|camping', keyword): grouped_concept = 'User: Camper'
        elif re.search(r'athlet|sport', keyword): grouped_concept = 'User: Athlete'
        elif re.search(r'tattoo', keyword): grouped_concept = 'User: Tattoo Care'
        elif re.search(r'pet|dog|cat', keyword): grouped_concept = 'User: Pet Owner'

    elif category == 'Format':
        if re.search(r'roll|roll of wipes', keyword): grouped_concept = 'Roll Format'
        elif re.search(r'individual|single pack|sachet', keyword): grouped_concept = 'Individual Pack Format'
        elif re.search(r'travel pack|portable|travel|on-the-go', keyword): grouped_concept = 'Travel Pack Format'
        elif re.search(r'bulk pack|value pack|bucket|large roll|high-count', keyword): grouped_concept = 'Bulk/Value Pack Format'
        elif re.search(r'dispenser|container', keyword): grouped_concept = 'Dispenser/Container Format'
        elif re.search(r'refill pack|rigid dispenser|peel-reseal|peel-and-reseal|rigid flip-top|hard flip-top', keyword): grouped_concept = 'Refill/peel-reseal/rigid flip-top Format'
        elif re.search(r'tub|canister|dispenser', keyword): grouped_concept = 'Tub/Canister Format'
        elif re.search(r'XL|extra large|extra-large', keyword): grouped_concept = 'XL size Format'
        elif re.search(r'pop-up tissue style', keyword): grouped_concept = 'pop-up tissue style Format'
        elif re.search(r'multi-pack|multi pack', keyword): grouped_concept = 'Multi-pack Format'
        elif re.search(r'small', keyword): grouped_concept = 'Small size Format'
        elif re.search(r'standard|medium|regular', keyword): grouped_concept = 'Standard/medium/regular size Format'
        elif re.search(r'extra-thick|extra thick', keyword): grouped_concept = 'Extra thick Format'
        elif re.search(r'soft|soft-pack|soft pack', keyword): grouped_concept = 'soft Format'
        elif re.search(r'smooth', keyword): grouped_concept = 'smooth Format'
        elif re.search(r'embossed|quilted', keyword): grouped_concept = 'embossed/quilted Format'
        elif re.search(r'box', keyword): grouped_concept = 'Box'

    return grouped_concept



# Apply the function to create the 'Grouped_Concept' column
exploded_df_with_concept = exploded_df.copy()
# Ensure the 'Keyword' column is available for the function
exploded_df_with_concept['Keyword'] = exploded_df_with_concept['Keyword'].astype(str)
exploded_df_with_concept['Grouped_Concept'] = exploded_df_with_concept.apply(group_keyword_concept, axis=1)

# Filter for high-volume keywords (Avg. monthly searches >= 500)
high_volume_keywords_df = exploded_df_with_concept[exploded_df_with_concept['Avg. monthly searches'] >= 500].copy()

# Aggregate high-volume keywords
aggregated_high_volume_concept = high_volume_keywords_df.groupby(['Category', 'Grouped_Concept']).agg(
    Total_Monthly_Searches=('Avg. monthly searches', 'sum'),
    Keyword_Count=('Keyword', 'nunique')
).reset_index()

# Display the head of the aggregated DataFrame
print("\nHead of aggregated_high_volume_concept:")
print(aggregated_high_volume_concept.head())

# Display the shape of the aggregated DataFrame
print(f"\nShape of aggregated_high_volume_concept: {aggregated_high_volume_concept.shape}")

# Load the 'all_classified_keywords.csv' file
# file_path = '/content/all_classified_keywords.csv'
# df_all_keywords = pd.read_csv(file_path)

# Apply the new group_keyword_concept function
# df_all_keywords['Grouped_Concept'] = df_all_keywords.apply(group_keyword_concept, axis=1)

# # Display the first few rows of the updated DataFrame
# print("First 5 rows of df_all_keywords with Grouped_Concept column:")
# display(df_all_keywords.head())

# # Display the shape of the updated DataFrame
# print(f"\nShape of df_all_keywords: {df_all_keywords.shape}")

# # Save the updated DataFrame to a new CSV file
# output_filename = '/content/all_classified_keywords_with_grouped_concept.csv'
# df_all_keywords.to_csv(output_filename, index=False)
# print(f"\nUpdated DataFrame saved to: {output_filename}")

# try:
#     files.download(output_filename)
#     print("Updated file downloaded.")
# except Exception as e:
#     print(f"Could not download updated file: {e}")
````

## Identify and Load Files

### Subtask:
List all CSV files within the specified Google Drive folder and load each into a pandas DataFrame, using the second row as the header.



**Reasoning**:
To identify and load all CSV files from the specified Google Drive folder, I will use the `os` module to list files and `pandas` to read each CSV, ensuring the second row is used as the header as instructed.



````python
import os
import pandas as pd

# Define the full path to the Google Drive folder
folder_path = '/content/drive/My Drive/Personal Care Project/Amazon Wipes Q1 2024 to Q4 2025'

# Get a list of all files in the specified folder
all_files = os.listdir(folder_path)

# Filter to include only CSV files
csv_files = [f for f in all_files if f.endswith('.csv')]

# Initialize an empty list to store the DataFrames
df_list_raw = []

print(f"Found {len(csv_files)} CSV files in '{folder_path}'.")

# For each CSV file, load it into a DataFrame using the second row as header
for file_name in csv_files:
    file_path = os.path.join(folder_path, file_name)
    try:
        # skiprows=1 makes the second row the header (0-indexed)
        df = pd.read_csv(file_path, skiprows=1)
        df_list_raw.append(df)
        print(f"Successfully loaded '{file_name}'.")
    except Exception as e:
        print(f"Error loading '{file_name}': {e}")

print(f"Total DataFrames loaded: {len(df_list_raw)}")
````

## Select and Combine Columns

### Subtask:
From each loaded DataFrame, select only the columns: "Search Frequency Rank", "Search Term", "Top Clicked Brand #1", "Top Clicked Brands #2", and "Top Clicked Brands #3". Then, combine all these processed DataFrames into a single, comprehensive DataFrame.


# Task
Prepare the 'Search Term' values from the `combined_df` DataFrame for LLM classification by extracting all unique 'Search Term' values.

**Reasoning**:
To select the required columns from each raw DataFrame and combine them into a single DataFrame, I will iterate through the list of raw DataFrames, select the specified columns, and then concatenate them. I will also handle cases where a column might be missing from a DataFrame.

````python
import pandas as pd

# Define the target columns
target_columns = [
    "Search Frequency Rank",
    "Search Term",
    "Top Clicked Brand #1",
    "Top Clicked Brands #2",
    "Top Clicked Brands #3",
    "Top Clicked Product #1: Product Title",
    "Top Clicked Product #2: Product Title",
    "Top Clicked Product #3: Product Title"

]

# Initialize an empty list to store DataFrames with selected columns
df_selected_columns_list = []

# Iterate through each raw DataFrame, select target columns, and append to the list
for i, df in enumerate(df_list_raw):
    # Ensure the columns exist before selecting or reindex to add missing columns as NaN
    # Using .reindex() is robust as it will add missing columns with NaN values
    selected_df = df.reindex(columns=target_columns)
    df_selected_columns_list.append(selected_df)
    print(f"Processed DataFrame {i+1}/{len(df_list_raw)}: Selected columns.")

# Concatenate all DataFrames in the list into a single DataFrame
combined_df = pd.concat(df_selected_columns_list, ignore_index=True)

# Define the output file path for the combined CSV
combined_filename = "/content/amzon_all_classified_keywords.csv"

# Save the combined DataFrame to a new CSV file
combined_df.to_csv(combined_filename, index=False)

print(f"All classified batch files combined into: {combined_filename}")
print(f"Total rows in combined file: {len(combined_df)}")

print(f"Successfully combined {len(df_selected_columns_list)} DataFrames into a single DataFrame with {len(combined_df)} rows and {len(combined_df.columns)} columns.")

# Display the head of the combined DataFrame
print("Combined DataFrame Head:")
print(combined_df.head())
````

## Apply Keyword Grouping to Top Clicked Brands Data

### Subtask:
Load the `amazon_top_clicked_brands_with_counts.csv` file, apply the `group_keyword_concept` function to group brand variations, re-aggregate the counts for these grouped brands, and then save and download the resulting CSV.

````python
import pandas as pd
from google.colab import files

# Load the amazon_top_clicked_brands_with_counts.csv file
file_path = '/content/amazon_top_clicked_brands_with_counts.csv'
brand_counts_df = pd.read_csv(file_path)

# Display the head of the loaded DataFrame
print("Original amazon_top_clicked_brands_with_counts.csv head:")
print(brand_counts_df.head())

# Prepare the DataFrame for the group_keyword_concept function
# The function expects 'Keyword' and 'Category' columns.
# Create a temporary DataFrame with these column names and a dummy 'Category' (Brand)
prepared_df = brand_counts_df.copy()
prepared_df.rename(columns={'Brand Name': 'Keyword'}, inplace=True)
prepared_df['Category'] = 'Brand' # Assign all these to the 'Brand' category for grouping logic

# Apply the group_keyword_concept function
# (Assuming group_keyword_concept is already defined in the environment)
prepared_df['Grouped_Concept'] = prepared_df.apply(group_keyword_concept, axis=1)

# Re-aggregate the data by 'Grouped_Concept' and sum the 'Count'
# Also count the unique original brand names that form each grouped concept
grouped_brand_counts = prepared_df.groupby('Grouped_Concept').agg(
    Total_Count=('Count', 'sum'),
    Unique_Brand_Names_Count=('Keyword', 'nunique')
).reset_index()

# Sort the grouped brand counts by Total_Count in descending order
grouped_brand_counts = grouped_brand_counts.sort_values(by='Total_Count', ascending=False)

print("\nGrouped Top Clicked Brands head:")
display(grouped_brand_counts.head(20))

# Define the output file path
output_filename = '/content/amazon_top_clicked_brands_with_counts_grouped.csv'

# Save the grouped DataFrame to a new CSV file
grouped_brand_counts.to_csv(output_filename, index=False)

print(f"\nGrouped top clicked brands with counts saved to: {output_filename}")

# Download the created CSV file
try:
    files.download(output_filename)
    print("Grouped top clicked brands with counts file downloaded.")
except Exception as e:
    print(f"Could not download grouped top clicked brands with counts file: {e}")
````

## Define and Run `classify_terms_with_llm` Function

````python
import pandas as pd
from openai import OpenAI
from io import StringIO
import os
import csv
from google.colab import files
import shutil # Import shutil for removing directories

from google.colab import userdata
a = userdata.get('api_key')

# Initialize client with your API key
client = OpenAI(api_key=a)

def classify_terms_batch_llm(batch_df: pd.DataFrame, batch_id: int) -> pd.DataFrame:
    """
    Classifies a batch of terms (product titles) using the OpenAI API.

    Args:
        batch_df: DataFrame containing a 'Keyword' column with unique product titles to classify.
        batch_id: Identifier for the current batch.

    Returns:
        DataFrame with classified results ('Product Title', 'Category') or an empty DataFrame if an error occurs.
    """
    # Ensure only the 'Keyword' column is used for the prompt
    keywords_csv_content = batch_df[['Keyword']].to_csv(index=False)

    prompt = f"""
    You are an expert in wipes, flushable wipes, XL wipes, Bamboo wipes consumer insights.
    Classify *all* of the following keywords into one or more of these categories.
    Do not skip any keywords, regardless of their search volume.
    If a keyword belongs to multiple categories, list them separated by a pipe symbol "|".


    **Predefined Categories:**
    - Brand
    - Claim
    - Format
    - Ingredient
    - Function/Benefit
    - Variants/Types
    - Scent
    - User

**Output Format:**
For each product title, return a CSV row with exactly two columns, including a header row:
Product Title,Category

**Product Titles to Classify (CSV format, include header row):**
Product Title
{keywords_csv_content}
Respond only with the CSV content, no additional text or explanations."""

    print(f"Sending batch {batch_id} with {len(batch_df)} product titles to OpenAI for classification...")
    try:
        response = client.chat.completions.create(
            model="gpt-4o",
            temperature=0, # Changed to 0 for more deterministic output.
            messages=[{"role": "user", "content": prompt}],
            response_format={"type": "text"}
        )
        output_text = response.choices[0].message.content

        # Robustly extract CSV content, handling markdown code blocks
        csv_content = output_text.strip()
        if '```csv' in csv_content:
            csv_content = csv_content.split('```csv', 1)[1].split('```', 1)[0].strip()
        elif '```plaintext' in csv_content:
            csv_content = csv_content.split('```plaintext', 1)[1].split('```', 1)[0].strip()
        elif '```' in csv_content:
            # Fallback for generic code block if type not specified
            parts = csv_content.split('```', 1)
            if len(parts) > 1:
                csv_content = parts[1].split('```', 1)[0].strip()
            else:
                csv_content = csv_content # No valid code block found or malformed

        # Attempt to parse as CSV with pandas first
        try:
            classified_df = pd.read_csv(StringIO(csv_content))
            # Validate columns
            if not all(col in classified_df.columns for col in ['Product Title', 'Category']):
                print(f"Warning: Batch {batch_id} - CSV header mismatch or missing columns with pandas. Trying parsing without header.")
                classified_df = pd.read_csv(StringIO(csv_content), header=None, names=['Product Title', 'Category'])
                # Check for NaNs that might indicate parsing issues even with header=None
                if classified_df.isnull().any().any() or classified_df.shape[1] != 2:
                    raise ValueError("Failed to parse cleanly with pandas.") # Trigger manual fallback
            return classified_df
        except (pd.errors.EmptyDataError, ValueError, pd.errors.ParserError) as csv_err:
            print(f"Error parsing CSV output for batch {batch_id} with pandas: {csv_err}")
            print(f"Attempting manual CSV parsing with csv.reader. Output text (first 500 chars): \n{csv_content[:500]}...")

            # Manual parsing fallback using csv.reader for robustness
            reader = csv.reader(StringIO(csv_content))
            data_rows = []
            header_skipped = False
            for i, row in enumerate(reader):
                if not row: # Skip empty rows
                    continue
                # More flexible header check (case-insensitive and whitespace trimmed)
                if not header_skipped and i == 0 and len(row) == 2 and row[0].strip().lower() == 'product title' and row[1].strip().lower() == 'category':
                    header_skipped = True
                    continue
                if len(row) == 2:
                    data_rows.append([item.strip() for item in row])
                else:
                    print(f"Warning: Batch {batch_id} - Skipping malformed row during manual parse: {row}")

            if data_rows:
                return pd.DataFrame(data_rows, columns=['Product Title', 'Category'])
            else:
                print(f"Error: Batch {batch_id} - Manual parsing with csv.reader also failed to extract data. Output: {csv_content}")
                return pd.DataFrame(columns=['Product Title', 'Category']) # Return empty DF

    except Exception as e:
        print(f"Error calling OpenAI API for batch {batch_id}: {e}")
        return pd.DataFrame(columns=['Product Title', 'Category'])

# --- Clear previous batches to ensure a clean run ---
output_dir = "/content/amazon_classified_batches/"
if os.path.exists(output_dir):
    shutil.rmtree(output_dir)
os.makedirs(output_dir, exist_ok=True)


# Load the unique product titles for LLM classification
product_titles_file_path = '/content/unique_amazon_product_titles_list.csv'
product_titles_df_for_llm = pd.read_csv(product_titles_file_path)

# IMPORTANT: Select ONLY the 'Unique Product Title' column before renaming to 'Keyword'
# This ensures the LLM prompt receives only the product title, not the count column.
if 'Unique Product Title' in product_titles_df_for_llm.columns:
    product_titles_df_for_llm = product_titles_df_for_llm[['Unique Product Title']]
    product_titles_df_for_llm.rename(columns={'Unique Product Title': 'Keyword'}, inplace=True)
elif 'Keyword' not in product_titles_df_for_llm.columns:
    print("Error: Neither 'Unique Product Title' nor 'Keyword' column found. Please check data source.")
    product_titles_df_for_llm = pd.DataFrame(columns=['Keyword'])

# --- Implement Batch Processing ---
batch_size = 500  # Define your desired batch size
classified_results_list = []

num_batches = (len(product_titles_df_for_llm) + batch_size - 1) // batch_size
print(f"Processing {len(product_titles_df_for_llm)} product titles in {num_batches} batches of size {batch_size}.")
print(f"Starting classification from batch 1...") # Always start from batch 1 after clearing directory

for i in range(num_batches):
    start_idx = i * batch_size
    end_idx = min((i + 1) * batch_size, len(product_titles_df_for_llm))
    batch_df = product_titles_df_for_llm.iloc[start_idx:end_idx]

    if not batch_df.empty:
        classified_batch = classify_terms_batch_llm(batch_df, i + 1)
        if not classified_batch.empty:
            classified_results_list.append(classified_batch)
            # Save each batch to a CSV file
            batch_filename = os.path.join(output_dir, f"amazon_classified_product_titles_batch_{i+1}.csv")
            classified_batch.to_csv(batch_filename, index=False)
            print(f"Batch {i+1} classified and saved to {batch_filename}")

            # Download each batch CSV to the user's local machine
            try:
                files.download(batch_filename)
                print(f"Batch {i+1} downloaded.")
            except Exception as e:
                print(f"Could not download batch {i+1} (may be due to Colab environment or browser settings): {e}")
        else:
            print(f"Batch {i + 1} returned empty classified results.")
    else:
        print(f"Skipping empty batch {i + 1}.")

# Concatenate all classified batches
all_completed_batches = []
for f in os.listdir(output_dir):
    if f.startswith('amazon_classified_product_titles_batch_') and f.endswith('.csv'):
        all_completed_batches.append(pd.read_csv(os.path.join(output_dir, f)))

if all_completed_batches:
    classified_search_df_stacked = pd.concat(all_completed_batches, ignore_index=True)
    print("\nAll batches classified and concatenated.")
else:
    classified_search_df_stacked = pd.DataFrame(columns=['Product Title', 'Category'])
    print("\nNo product titles were classified or found in saved batches.")


# Display the first few rows of the classified DataFrame
print("\nFirst 5 rows of classified_search_df_stacked:")
print(classified_search_df_stacked.head())

# Display the shape of the classified DataFrame
print(f"\nShape of classified_search_df_stacked: {classified_search_df_stacked.shape}")
````

## Explode Search Term Classifications

### Subtask:
Handle any search terms that are classified into multiple categories (e.g., 'Category1|Category2'). These will be split into separate rows, with each row representing one category for that search term.


**Reasoning**:
To handle multiple classifications for search terms, I will convert the 'Category' column to string, split it by the pipe symbol, explode the DataFrame, and then clean up any whitespace.



````python
import pandas as pd

# First, ensure 'Product Title' and 'Category' columns are string type for robust comparison
classified_search_df_stacked['Product Title'] = classified_search_df_stacked['Product Title'].astype(str)
classified_search_df_stacked['Category'] = classified_search_df_stacked['Category'].astype(str)

# Remove all rows that might be redundant headers (e.g., 'Product Title', 'Category')
# This is a more robust way to clean if multiple batches inserted headers
initial_rows = len(classified_search_df_stacked)
classified_search_df_stacked = classified_search_df_stacked[
    ~((classified_search_df_stacked['Product Title'].str.strip() == 'Product Title') &
      (classified_search_df_stacked['Category'].str.strip().isin(['Category', "['Category']", "'Category'"])))
].copy()

if len(classified_search_df_stacked) < initial_rows:
    print(f"Removed {initial_rows - len(classified_search_df_stacked)} redundant header rows from classified_search_df_stacked.")

# Clean the 'Category' column: remove leading/trailing spaces, brackets, quotes, and backslashes
classified_search_df_stacked['Category'] = classified_search_df_stacked['Category'].str.strip()
classified_search_df_stacked['Category'] = classified_search_df_stacked['Category'].str.replace(r"[\[\]'\"\\]", "", regex=True)

# Replace comma with pipe for consistency in multi-category entries, then split by pipe
# This assumes that if a comma is present, it's intended to separate categories.
classified_search_df_stacked['Category'] = classified_search_df_stacked['Category'].str.replace(',', '|', regex=False)
classified_search_df_stacked['Category'] = classified_search_df_stacked['Category'].str.split('|')

# Explode the DataFrame to create a new row for each category
exploded_classifications_df = classified_search_df_stacked.explode('Category')

# Remove any leading or trailing whitespace from the 'Category' column after exploding
exploded_classifications_df['Category'] = exploded_classifications_df['Category'].str.strip()

# Display the first few rows of the exploded DataFrame
print("\nFirst 5 rows of exploded_classifications_df:")
print(exploded_classifications_df.head())

# Display the shape of the exploded DataFrame
print(f"\nShape of exploded_classifications_df: {exploded_classifications_df.shape}")
````

# Task
Load and merge the Amazon data from "amazon_exploded_classifications.csv" and "unique_amazon_product_titles_list.csv" to create a single DataFrame that associates product titles with their classification categories and their corresponding product title counts.

## Load and Merge Amazon Data

### Subtask:
Load `amazon_exploded_classifications.csv` and `unique_amazon_product_titles_list.csv`. Then, merge these two DataFrames to associate each product title with its classification categories and its count, which will serve as the 'volume' metric for Amazon data.


**Reasoning**:
To begin the subtask, I will load the two specified CSV files into pandas DataFrames, `df_classifications` and `df_counts` respectively. This is the first instruction and sets up the data for subsequent merging.

## Filter High-Volume Concepts and Aggregate

### Subtask:
Filter the grouped Amazon data to include only 'high-volume' concepts, then aggregate the filtered data by 'Category' and 'Grouped_Concept'.


**Reasoning**:
To filter the grouped Amazon data for high-volume concepts and aggregate it, I will filter `amazon_grouped_df` by 'Product Title Count' >= 10, then group by 'Category' and 'Grouped_Concept', calculating the sum of 'Product Title Count' and the count of unique 'Keyword' entries.



````python
import pandas as pd

# 1. Filter the amazon_grouped_df DataFrame to include only rows where the 'Product Title Count' is greater than or equal to 10.
# Store the result in a new DataFrame called high_volume_amazon_df.
high_volume_amazon_df = amazon_grouped_df[amazon_grouped_df['Product Title Count'] >= 10].copy()

# 2. Group the high_volume_amazon_df by 'Category' and 'Grouped_Concept'.
# 3. For each group, calculate the sum of 'Product Title Count' and name this aggregated column 'Total_Product_Title_Count'.
# 4. For each group, also count the number of unique 'Keyword' entries (which represent the original product titles)
#    and name this aggregated column 'Unique_Product_Titles_Count'.
# 5. Store the resulting aggregated DataFrame in a variable named high_volume_grouped_concepts_summary.
high_volume_grouped_concepts_summary = high_volume_amazon_df.groupby(['Category', 'Grouped_Concept']).agg(
    Total_Product_Title_Count=('Product Title Count', 'sum'),
    Unique_Product_Titles_Count=('Keyword', 'nunique')
).reset_index()

# 6. Print the first 5 rows and the shape of the high_volume_grouped_concepts_summary DataFrame to verify the output.
print("\nFirst 5 rows of high_volume_grouped_concepts_summary:")
print(high_volume_grouped_concepts_summary.head())

print(f"\nShape of high_volume_grouped_concepts_summary: {high_volume_grouped_concepts_summary.shape}")
````

## Define Keyword Segment Extraction Function

### Subtask:
Create a new helper function, `get_matched_segment`, that takes the `original_keyword` (product title) and its `category` as input. This function will re-apply the regex patterns used in the `group_keyword_concept` function. For each match, it will return the exact substring from the `original_keyword` that caused the match. If no specific pattern from the grouping logic is found, it will return the full original keyword.


**Reasoning**:
To extract specific keyword segments, I will define a new helper function `get_matched_segment` that re-applies the regex patterns from the `group_keyword_concept` function to the original product titles. This function will return the exact matched substring or the full original keyword if no specific pattern is found, handling 'Brand' category priorities.



````python
import re

def get_matched_segment(original_keyword: str, category: str) -> str:
    """
    Extracts the exact substring from the original_keyword that matches a regex pattern
    based on its assigned category. If no specific pattern is matched, it returns
    the full original_keyword.
    """
    keyword_lower = original_keyword.lower()

    # Helper to check for a match and return the matched segment
    def find_and_return_match(pattern, original_kw):
        match = re.search(pattern, original_kw, re.IGNORECASE)
        if match: return match.group(0)
        return None

    if category == 'Brand':
        # Prioritize Private Label Brands
        private_label_brands_pattern = r'walmart|365|equate|parent\s?s choice|target|up & up|cvs|cvs health|cvs natural baby wipes|cvs pure water baby wipes|walgreens|costco|kirkland signature|kirkland|sam|sam\s? club|member\s?s mark|kroger|comforts|publix|aldi|little journey|lidl|lupilu|meijer|h-e-b|albertsons|safeway|signature care|dollar general|dg baby|family dollar|kidgets|family wellness|bj\s?s wholesale|berkley jensen|shoppers drug mart|life brand superstore|loblaws|tesco|fred & flo|sainsbury\s?s|little ones|asda|little angels|morrisons|nutmeg'
        match = find_and_return_match(private_label_brands_pattern, original_keyword)
        if match: return match

        # Prioritize Amazon Private Label Brands
        amazon_private_label_pattern = r'amazon|mama bear|amazon elements|Amazon Elements'
        match = find_and_return_match(amazon_private_label_pattern, original_keyword)
        if match: return match

        # Other common wipe brands
        brand_patterns = {
            'huggies': r'huggies',
            'honest': r'honest',
            'charmin': r'charmin',
            'kandoo': r'kandoo',
            'cottonelle': r'cottonelle',
            'pampers': r'pampers',
            'waterwipes': r'waterwipes',
            'wet ones': r'wet ones',
            'seventh generation': r'seventh generation',
            'dyper': r'dyper',
            "nice 'n clean": r"nice 'n clean|nice n clean|nice and clean",
            'boogie wipes': r'boogie wipes',
            'duke cannon': r'duke cannon',
            'goodwipes': r'goodwipes',
            'action wipes': r'action wipes',
            'stall mates': r'stall mates',
            'dude wipes': r'dude wipes',
            'waterwipe|waterwipes': r'waterwipe|waterwipes',
            'curash': r'curash',
            'bumboosa': r'bumboosa',
            'neosporin': r'neosporin',
            'kimberly-clark': r'kimberly-clark',
            'the cheeky panda': r'the cheeky panda'
        }
        for pattern in brand_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Claim':
        claim_patterns = {
            'flushable': r'flushable|dispersible|flush',
            'biodegradable': r'biodegradable|compostable',
            'sensitive/hypoallergenic': r'sensitive|hypoallergenic',
            'unscented/fragrance free': r'unscented|fragrance free|fragrance-free|without fragrance|scent free|scent-free',
            'natural/organic': r'natural|organic',
            'dermatologist tested': r'dermatologist tested',
            'alcohol-free': r'alcohol free',
            'paraben-free': r'paraben free',
            'best/top': r'best|top',
            'eco-friendly': r'eco-friendly|eco',
            'plant-based': r'plant-based|plant',
            'durable/strong': r'durable|strong',
            'gentle': r'gentle',
            'pure': r'pure',
            'disposable': r'disposable',
            'ph balanced': r'ph balanced',
            'travel friendly': r'travel friendly',
            'bulk purchase': r'bulk'
        }
        for pattern in claim_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Ingredient':
        ingredient_patterns = {
            'water': r'water',
            'aloe vera': r'aloe vera|aloe',
            'bamboo': r'bamboo',
            'cotton': r'cotton',
            'shea butter': r'shea butter',
            'vitamin e': r'vitamin e',
            'glycerin': r'glycerin',
            'chamomile': r'chamomile',
            'plant fiber': r'plant fiber|plant-fiber',
            'essential oil': r'essential oil'
        }
        for pattern in ingredient_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Function/Benefit':
        function_patterns = {
            'cleaning/cleansing': r'cleaning|cleanse',
            'personal/feminine hygiene': r'personal hygiene|feminine hygiene',
            'toilet use': r'toilet|after toilet',
            'freshening/refreshing': r'freshening|refreshing',
            'moisturizing/hydrating': r'moisturizing|hydrating',
            'makeup remover': r'makeup remover',
            'antiseptic/antibacterial': r'antiseptic|antibacterial',
            'disinfecting': r'disinfecting',
            'odor control': r'odor control',
            'travel': r'travel',
            'outdoor use': r'outdoor',
            'adult care': r'adult care',
            'incontinence care': r'incontinence',
            'hemorrhoid care': r'hemorrhoid|hemorrhoidal',
            'first aid': r'first aid'
        }
        for pattern in function_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Variants/Types':
        variants_patterns = {
            'baby wipes': r'baby wipes|infant wipes',
            'face wipes': r'face wipes|facial wipes',
            'body wipes': r'body wipes',
            'toilet wipes': r'toilet wipes',
            'feminine/intimate wipes': r'feminine wipes|intimate wipes',
            'adult wipes': r'adult wipes',
            'camping/travel wipes': r'camping wipes|travel wipes',
            'gym/sport wipes': r'gym wipes|sport wipes',
            'flushable wipes': r'flushable wipes',
            'wet wipes': r'wet wipes',
            'dry wipes': r'dry wipes',
            'medicated wipes': r'medicated wipes',
            'anti-itch wipes': r'anti-itch wipes',
            'pet wipes': r'dog wipes|pet wipes',
            'hand wipes': r'hand wipes',
            'antibacterial wipes': r'antibacterial wipes',
            'xl wipes': r'xxl|xl wipes',
            'personal wipes': r'personal wipes'
        }
        for pattern in variants_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Scent':
        scent_patterns = {
            'unscented': r'unscented|fragrance free',
            'fresh scent': r'fresh scent',
            'coconut scent': r'coconut scent',
            'lavender scent': r'lavender scent',
            'citrus scent': r'citrus scent'
        }
        for pattern in scent_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'User':
        user_patterns = {
            'user: baby': r'baby|infant',
            'user: kids/children': r'kids|children|child',
            'user: adult': r'adult|adults',
            'user: men': r'men',
            'user: women/feminine': r'women|feminine',
            'user: elderly/senior': r'elderly|senior',
            'user: patients/incontinence': r'patient|incontinence',
            'user: diaper rash sufferers': r'diaper rash',
            'user: traveler': r'traveler|travel',
            'user: camper': r'camper|camping',
            'user: athlete': r'athlet|sport',
            'user: tattoo care': r'tattoo',
            'user: pet owner': r'pet|dog|cat'
        }
        for pattern in user_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Format':
        format_patterns = {
            'roll format': r'roll|roll of wipes',
            'individual pack format': r'individual|single pack|sachet',
            'travel pack format': r'travel pack|portable|travel|on-the-go',
            'bulk/value pack format': r'bulk pack|value pack|bucket|large roll|high-count',
            'dispenser/container format': r'dispenser|container',
            'refill/peel-reseal/rigid flip-top format': r'refill pack|rigid dispenser|peel-reseal|peel-and-reseal|rigid flip-top|hard flip-top',
            'tub/canister format': r'tub|canister|dispenser',
            'xl size format': r'XL|extra large|extra-large',
            'pop-up tissue style format': r'pop-up tissue style',
            'multi-pack format': r'multi-pack|multi pack',
            'small size format': r'small',
            'standard/medium/regular size format': r'standard|medium|regular',
            'extra thick format': r'extra-thick|extra thick',
            'soft format': r'soft|soft-pack|soft pack',
            'smooth format': r'smooth',
            'embossed/quilted format': r'embossed|quilted',
            'box': r'box'
        }
        for pattern in format_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    # If no specific pattern matches within the category, return the original keyword
    return original_keyword
````

**Reasoning**:
The `get_matched_segment` function has been defined. Now, I will apply this function to the `high_volume_amazon_df_for_extraction` DataFrame to create the `Extracted_Keyword_Segment` column, then select the required columns and save the result to a new CSV file.



````python
import pandas as pd
from google.colab import files

# Apply the get_matched_segment function to the DataFrame
high_volume_amazon_df_for_extraction['Extracted_Keyword_Segment'] = high_volume_amazon_df_for_extraction.apply(
    lambda row: get_matched_segment(row['Keyword'], row['Category']),
    axis=1
)

# Select the required columns for the final output
final_output_df = high_volume_amazon_df_for_extraction[[
    'Category',
    'Grouped_Concept',
    'Extracted_Keyword_Segment'
]].copy()

# Define the output file path
output_filename = "/content/amazon_grouped_concept_keywords.csv"

# Save the DataFrame to a new CSV file
final_output_df.to_csv(output_filename, index=False)

print(f"Generated file saved to: {output_filename}")

# Display the first 5 rows of the generated DataFrame
print("\nFirst 5 rows of the generated CSV file:")
print(final_output_df.head())

# Summarize the generated file and extraction process
print("\nSummary of generated file and extraction process:")
print(f"The file '{output_filename}' contains {len(final_output_df)} rows and {len(final_output_df.columns)} columns.")
print("It includes the 'Category', 'Grouped_Concept', and 'Extracted_Keyword_Segment' for high-volume Amazon product titles.")
print("The 'Extracted_Keyword_Segment' column is derived by reapplying the regex patterns from the grouping logic to the original product titles, capturing the exact matching substring.")

# Download the created CSV file
try:
    files.download(output_filename)
    print("Generated file downloaded successfully.")
except Exception as e:
    print(f"Could not download generated file: {e}")
````

To address the identified improvements, I will update the `get_matched_segment` function in the specified cell with expanded regex patterns for brands, formats, claims, and function/benefits to ensure more granular keyword extraction.



````python
import re

def get_matched_segment(original_keyword: str, category: str) -> str:
    """
    Extracts the exact substring from the original_keyword that matches a regex pattern
    based on its assigned category. If no specific pattern is matched, it returns
    the full original_keyword.
    """
    keyword_lower = original_keyword.lower()

    # Helper to check for a match and return the matched segment
    def find_and_return_match(pattern, original_kw):
        match = re.search(pattern, original_kw, re.IGNORECASE)
        if match: return match.group(0)
        return None

    if category == 'Brand':
        # Prioritize Private Label Brands
        private_label_brands_pattern = r'walmart|365|equate|parent\s?s choice|target|up & up|cvs|cvs health|cvs natural baby wipes|cvs pure water baby wipes|walgreens|costco|kirkland signature|kirkland|sam|sam\s? club|member\s?s mark|kroger|comforts|publix|aldi|little journey|lidl|lupilu|meijer|h-e-b|albertsons|safeway|signature care|dollar general|dg baby|family dollar|kidgets|family wellness|bj\s?s wholesale|berkley jensen|shoppers drug mart|life brand superstore|loblaws|tesco|fred & flo|sainsbury\s?s|little ones|asda|little angels|morrisons|nutmeg'
        match = find_and_return_match(private_label_brands_pattern, original_keyword)
        if match: return match

        # Prioritize Amazon Private Label Brands
        amazon_private_label_pattern = r'amazon|mama bear|amazon elements|Amazon Elements'
        match = find_and_return_match(amazon_private_label_pattern, original_keyword)
        if match: return match

        # Other common wipe brands (expanded)
        brand_patterns = {
            'huggies': r'huggies',
            'honest': r'honest',
            'charmin': r'charmin',
            'kandoo': r'kandoo',
            'cottonelle': r'cottonelle',
            'pampers': r'pampers',
            'waterwipes': r'waterwipes',
            'wet ones': r'wet ones',
            'seventh generation': r'seventh generation',
            'dyper': r'dyper',
            "nice 'n clean": r"nice 'n clean|nice n clean|nice and clean",
            'boogie wipes': r'boogie wipes',
            'duke cannon': r'duke cannon',
            'goodwipes': r'goodwipes',
            'action wipes': r'action wipes',
            'stall mates': r'stall mates',
            'dude wipes': r'dude wipes',
            'waterwipe|waterwipes': r'waterwipe|waterwipes',
            'curash': r'curash',
            'bumboosa': r'bumboosa',
            'neosporin': r'neosporin',
            'kimberly-clark': r'kimberly-clark',
            'the cheeky panda': r'the cheeky panda',
            'dermoplast': r'dermoplast',
            'noveha': r'noveha',
            'clean skin club': r'clean skin club',
            'dapple baby': r'dapple baby',
            'pet md': r'pet md',
            'tucks': r'tucks',
            'tide': r'tide',
            'weiman': r'weiman',
            'vagisil': r'vagisil',
            'babyganics': r'babyganics',
            'combat wipes': r'combat wipes',
            'leqxgo': r'leqxgo',
            'archxx': r'archxx',
            'stridex': r'stridex',
            'meduwedu': r'meduwedu',
            'super fresh': r'super fresh',
            'med pride': r'med pride',
            'purefresh': r'purefresh',
            'windex': r'windex',
            'zeiss': r'zeiss',
            'oxo tot': r'oxo tot',
            'professional disposables': r'professional disposables',
            'cutter': r'cutter',
            'miraclewipes': r'miraclewipes',
            'care touch': r'care touch',
            'mightygood': r'mightygood',
            'hicc pet': r'hicc pet',
            'medela': r'medela',
            'best pet supplies': r'best pet supplies',
            'mygezi': r'mygezi',
            'impossibly compact': r'impossibly compact',
            'uni solve': r'uni-solve|unisovle',
            't.n. dickinson': r't.n. dickinson|tn dickinson',
            'earth rated': r'earth rated',
            'lume': r'lume',
            'curad': r'curad',
            'epielle': r'epielle',
            'arm & hammer': r'arm & hammer|arm and hammer',
            'vet’s best': r'vet’s best|vets best',
            'mimorou': r'mimorou',
            'sono': r'sono',
            'wags & wiggles': r'wags & wiggles|wags and wiggles',
            'momcozy': r'momcozy',
            'flenpptly': r'flenpptly',
            'neutrogena': r'neutrogena',
            'forpro professional collection': r'forpro professional collection'
        }
        for pattern in brand_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Claim':
        claim_patterns = {
            'flushable': r'flushable|dispersible|flush',
            'biodegradable': r'biodegradable|compostable',
            'sensitive/hypoallergenic': r'sensitive|hypoallergenic',
            'unscented/fragrance free': r'unscented|fragrance free|fragrance-free|without fragrance|scent free|scent-free',
            'natural/organic': r'natural|organic',
            'dermatologist tested': r'dermatologist tested',
            'alcohol-free': r'alcohol free',
            'paraben-free': r'paraben free',
            'best/top': r'best|top',
            'eco-friendly': r'eco-friendly|eco',
            'plant-based': r'plant-based|plant',
            'durable/strong': r'durable|strong',
            'gentle': r'gentle',
            'pure': r'pure',
            'disposable': r'disposable',
            'ph balanced': r'ph balanced',
            'travel friendly': r'travel friendly',
            'bulk purchase': r'bulk',
            'soothing': r'soothing',
            'sensitive eyes': r'sensitive eyes'
        }
        for pattern in claim_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Ingredient':
        ingredient_patterns = {
            'water': r'water',
            'aloe vera': r'aloe vera|aloe',
            'bamboo': r'bamboo',
            'cotton': r'cotton',
            'shea butter': r'shea butter',
            'vitamin e': r'vitamin e',
            'glycerin': r'glycerin',
            'chamomile': r'chamomile',
            'plant fiber': r'plant fiber|plant-fiber',
            'essential oil': r'essential oil'
        }
        for pattern in ingredient_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Function/Benefit':
        function_patterns = {
            'cleaning/cleansing': r'cleaning|cleanse|cleansing',
            'personal/feminine hygiene': r'personal hygiene|feminine hygiene',
            'toilet use': r'toilet|after toilet',
            'freshening/refreshing': r'freshening|refreshing',
            'moisturizing/hydrating': r'moisturizing|hydrating',
            'makeup remover': r'makeup remover',
            'antiseptic/antibacterial': r'antiseptic|antibacterial',
            'disinfecting': r'disinfecting',
            'odor control': r'odor control',
            'travel': r'travel',
            'outdoor use': r'outdoor',
            'adult care': r'adult care',
            'incontinence care': r'incontinence',
            'hemorrhoid care': r'hemorrhoid|hemorrhoidal',
            'first aid': r'first aid',
            'wound care': r'wound care',
            'rinse-free bathing': r'rinse-free bathing|showerless bath|waterless bath'
        }
        for pattern in function_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Variants/Types':
        variants_patterns = {
            'baby wipes': r'baby wipes|infant wipes',
            'face wipes': r'face wipes|facial wipes',
            'body wipes': r'body wipes',
            'toilet wipes': r'toilet wipes',
            'feminine/intimate wipes': r'feminine wipes|intimate wipes',
            'adult wipes': r'adult wipes',
            'camping/travel wipes': r'camping wipes|travel wipes',
            'gym/sport wipes': r'gym wipes|sport wipes',
            'flushable wipes': r'flushable wipes',
            'wet wipes': r'wet wipes',
            'dry wipes': r'dry wipes',
            'medicated wipes': r'medicated wipes',
            'anti-itch wipes': r'anti-itch wipes',
            'pet wipes': r'dog wipes|pet wipes',
            'hand wipes': r'hand wipes',
            'antibacterial wipes': r'antibacterial wipes',
            'xl wipes': r'xxl|xl wipes',
            'personal wipes': r'personal wipes'
        }
        for pattern in variants_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Scent':
        scent_patterns = {
            'unscented': r'unscented|fragrance free',
            'fresh scent': r'fresh scent',
            'coconut scent': r'coconut scent',
            'lavender scent': r'lavender scent',
            'citrus scent': r'citrus scent'
        }
        for pattern in scent_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'User':
        user_patterns = {
            'user: baby': r'baby|infant',
            'user: kids/children': r'kids|children|child',
            'user: adult': r'adult|adults',
            'user: men': r'men',
            'user: women/feminine': r'women|feminine',
            'user: elderly/senior': r'elderly|senior',
            'user: patients/incontinence': r'patient|incontinence',
            'user: diaper rash sufferers': r'diaper rash',
            'user: traveler': r'traveler|travel',
            'user: camper': r'camper|camping',
            'user: athlete': r'athlet|sport',
            'user: tattoo care': r'tattoo',
            'user: pet owner': r'pet|dog|cat'
        }
        for pattern in user_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    elif category == 'Format':
        format_patterns = {
            'roll format': r'roll|roll of wipes',
            'individual pack format': r'individual|single pack|sachet',
            'travel pack format': r'travel pack|portable|travel|on-the-go',
            'bulk/value pack format': r'bulk pack|value pack|bucket|large roll|high-count',
            'dispenser/container format': r'dispenser|container',
            'refill/peel-reseal/rigid flip-top format': r'refill pack|rigid dispenser|peel-reseal|peel-and-reseal|rigid flip-top|hard flip-top',
            'tub/canister format': r'tub|canister|dispenser',
            'xl size format': r'XL|extra large|extra-large',
            'pop-up tissue style format': r'pop-up tissue style',
            'multi-pack format': r'multi-pack|multi pack',
            'small size format': r'small',
            'standard/medium/regular size format': r'standard|medium|regular',
            'extra thick format': r'extra-thick|extra thick',
            'soft format': r'soft|soft-pack|soft pack',
            'smooth format': r'smooth',
            'embossed/quilted format': r'embossed|quilted',
            'box': r'box',
            # Added numerical counts and packaging descriptions
            'pack count': r'(\d+)\s*(?:pack|count|ct|wipes|total wipes)',
            'flip-top': r'flip-top',
            'resealable': r'resealable',
            'soft-pack': r'soft-pack|soft pack'
        }
        for pattern in format_patterns.values():
            match = find_and_return_match(pattern, original_keyword)
            if match: return match

    # If no specific pattern matches within the category, return the original keyword
    return original_keyword
````
